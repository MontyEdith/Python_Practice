```python
#Check whether a year is a leap year or not.

year = int(input("Enter the year: "))

if (year%400 == 0) or ((year%100 != 0) and (year%4 == 0)):
    print("Entered year is leap yeap")
else:
    print("Entered year is not leap yeap")
```
