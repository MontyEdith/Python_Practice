```python
#Create a simple calculator using +, -, *, and /.

num1 = float(input("Enter 1st number: "))
num2 = float(input("Enter 2nd number: "))
print("Enter proper symbol for the operation you want to proceed with")
print("For Addition: Enter '+' ")
print("For Substraction: Enter '-' ")
print("For Multiplication: Enter '*' ")
print("For Division: Enter '/' ")
oper = input("Enter your choice: ")

if oper == "+":
    print("\nAddition of entered nnumbers is:", num1+num2)
elif oper == "-":
    print("\nSubstraction of entered nnumbers is:", num1-num2)
elif oper == "*":
    print("\nMultiplication of entered nnumbers is:", num1*num2)
elif oper == "/":
    print("\nDivision of entered nnumbers is:", num1/num2)
else:
    print("\nInvalid Input")
```

```python
#Determine a person's BMI category using if-elif-else.

"""Underweight: Less than 18.5
Healthy Weight: 18.5 to less than 25
Overweight: 25 to less than 30
Obesity: 30 or greater"""

weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in m: "))

BMI = weight/(height)**2
print(BMI)

if BMI < 18.5:
    print("Underweight")
elif BMI >= 18.5 and BMI <25:
    print("Healthy Weight")
elif BMI >= 25 and BMI <30:
    print("Overweight")
elif BMI >= 30:
    print("Obesity")
else:
    print("\nInvalid Input")
```

Example Tiered Structure
Spend ₹1,000 to ₹2,499: Get 5% off your total purchase amount.
Spend ₹2,500 to ₹4,999: Get 10% off your total purchase amount.
Spend ₹5,000 or more: Get 15% off your total purchase amount.

```python
#Calculate discount based on purchase amount.

purchase_amt = float(input("Enter your purchase amount: "))

if purchase_amt >= 5000:
    print("Congrats, you '15%' off your total purchase amount and you discount amount on your purchase:", purchase_amt*0.15)
elif purchase_amt >= 2500 and purchase_amt < 5000:
    print("Congrats, you '10%' off your total purchase amount and you discount amount on your purchase:", purchase_amt*0.10)
elif purchase_amt >= 1000 and purchase_amt < 2500:
    print("Congrats, you '5%' off your total purchase amount and you discount amount on your purchase:", purchase_amt*0.05)
else:
    print("Sorry, No discount for you")
```

```python
#Determine the type of triangle based on its sides: equilateral, isosceles, or scalene.

side1 = int(input("Enter side1: "))
side2 = int(input("Enter side2: "))
side3 = int(input("Enter side3: "))

if side1 == side2 and side1 == side3:
    print("Triangle is equilateral")
elif side1 == side2 and side1 != side3:
    print("Triangle is isosceles")
elif side1 != side2 and side1 == side3:
    print("Triangle is isosceles")
elif side1 != side2 and side2 == side3:
    print("Triangle is isosceles")
else:
    print("Triangle is scalene")
```

```python
#Check whether a person is eligible to vote and, if eligible, check whether they have a valid voter ID.

age = int(input("Enter your age: "))

if age >= 18:
    voter_ID = input("Do you have valid voter ID? (y/n)")
    if voter_ID == "y":
        print("You are eligible for vote with valid voter ID")
    else:
        print("You are eligible for vote but you don't have valid voter ID")
else:
    print("You are not eligible for vote")
```

```python
#Check whether a student passed and then determine their grade.

marks = float(input("Enter your marks: "))

if marks >= 40:
    if marks >= 90 and marks <=100:
        print("You have passed the exam and scored A")
    elif marks >= 80 and marks <=89:
        print("You have passed the exam and scored B")
    elif marks >= 70 and marks <=79:
        print("You have passed the exam and scored C")
    elif marks >= 60 and marks <=69:
        print("You have passed the exam and scored D")
    elif marks >= 40 and marks <60:
        print("You have passed the exam and scored E")
    else:
        print("Invalid Input")
else:
    print("You have failed the exam")
    

```

```python
#Check whether a number is positive and then determine whether it is even or odd.

num = float(input("Enter any number: "))

if num > 0:
    if num%2 ==0:
        print("Entered number is Positive even")
    else:
        print("Entered number is Positive odd")
else:
    print("Entered number is Non-positive number")
```

```python
#Check whether a person is old enough to drive and whether they have a valid license.

age = int(input("Enter your age: "))

if age >= 18:
    license = input("Do you have valid license? (y/n)")
    if license == "y":
        print("You are old enough to drive with valid license")
    else:
        print("You are old enough to drive but you don't have valid license")
else:
    print("You are not old enough to drive")
```

```python
#Check whether a number is divisible by 5 and then check whether it is also divisible by 10.

num = float(input("Enter any number: "))

if num%5==0:
    if num%10 ==0:
        print("Entered number is divisible by both 5 and 10")
    else:
        print("Entered number is divisible by 5 but not by 10")
else:
    print("Entered number is not divisible by 5")
```

```python
#Check whether a student passed all three subjects before calculating their average.

eng = int(input("Enter your english marks: "))
sst = int(input("Enter your social science marks: "))
math = int(input("Enter your maths marks: "))

if eng >= 40:
    if sst >= 40:
        if math >= 40:
            print("Students have passes all three subjects ")
            avg = (eng+sst+math)/3
            print("Average:", avg)
        else:
            print("Students failed maths")
    else:
        print("Students failed social science")
else:
    print("Students failed english")

```

```python
#Check whether an ATM withdrawal is valid and then check whether sufficient balance exists.

atm_pin = "1234"
balance = 1000

pin = input("Enter you ATM pin: ")

if pin == atm_pin:
    print("Welcome to the ABC Bank\n")
    withdrawal_amt = int(input("Enter the amount you want to withdraw: "))
    if withdrawal_amt <= balance:
        print("You have suffient balance to do the withdrawal")
    else:
        print("You don't have suffient balance to do the withdrawal")
else:
    print("Incorrect Pin")
```

```python
#Check whether a product is available and then check whether the requested quantity is in stock.

product_available = input("Confirm whether a product is available? (y/n)")
stock = 10

if product_available == "y":
    requested_quatity = int(input("Enter the your requested quantity: "))
    if requested_quatity <=stock:
        print("Product is available and the requested quantity is in stock")
    else:
        print("Product is available and the requested quantity is not in stock")
else:
    print("Product is not available")
```
