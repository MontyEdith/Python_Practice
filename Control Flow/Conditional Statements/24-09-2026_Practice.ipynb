```python
"""Create an ATM withdrawal program that checks:
Minimum withdrawal amount
Available balance
Whether the amount is a multiple of 100"""

print("Welcome to the XYZ Bank")
min_withdrawal_amt = 500
avail_bal = 2000

withdrawal_amt = int(input("Enter the amount you want to withdraw: "))
print(withdrawal_amt)

if withdrawal_amt > min_withdrawal_amt:
    if withdrawal_amt <= avail_bal:
        if withdrawal_amt%100 == 0:
            print("You can do the withdrawal")
        else:
            print("Please enter the withdrawal amount in a multiple of 100")
    else:
        print("You don't have enough balance")
else:
        print("The amount you want to withdraw is less than minimum withdrawal amount")
        
```

```python
#Determine whether a triangle is right-angled, acute, or obtuse.

side1 = float(input("Enter 1st side: "))
side2 = float(input("Enter 2nd side: "))
side3 = float(input("Enter 3rd side: "))

print(side1, side2, side3)

if side1 > side2 and side1 > side3:
    if (side1**2) == (side2**2 + side3**2):
        print("As per the entered sides, triangle is right-angled")
    elif (side1**2) < (side2**2 + side3**2):
            print("As per the entered sides, triangle is acute")
    elif (side1**2) > (side2**2 + side3**2):
            print("As per the entered sides, triangle is obtuse")
    else:
        print("Invalid input")
else:
        print("Invalid input1")
```

```python
#Determine whether a triangle is right-angled, acute, or obtuse.

side1 = float(input("Enter 1st side: "))
side2 = float(input("Enter 2nd side: "))
side3 = float(input("Enter 3rd side: "))

print(side1, side2, side3)

#Checking whether sides can form a triangle
if ((side1+side2) > side3) and ((side2+side3) > side1) and ((side1+side3) > side2):  
    
    #Checking the largest side
    if side1>side2 and side1>side3:
        largest_side = side1
        smallest_side1 = side2
        smallest_side2 = side3
    elif side2>side1 and side2>side3:
        largest_side = side2
        smallest_side1 = side1
        smallest_side2 = side3
    else:
        largest_side = side3
        smallest_side1 = side1
        smallest_side2 = side2

    #Checking whether a triangle is right-angled, acute, or obtuse
    if (largest_side**2) == (smallest_side1**2 + smallest_side2**2):
        print("As per the entered sides, triangle is right-angled")
    elif (largest_side**2) < (smallest_side1**2 + smallest_side2**2):
        print("As per the entered sides, triangle is acute")   
    else:
        print("As per the entered sides, triangle is obtuse")
else:
    print("Given sides can't form a valid triangle")
```

```python
#Create a simple banking system that allows withdrawal only when the account is active and the balance is sufficient.

balance = 2000
account = input("Enter your account status?(active/inactive)")

if account == "active":
    withdrawal_amt = int(input("Enter your withdrawal amount: "))
    if withdrawal_amt <= balance:
        print("You can do the withdrawal")
    else:
        print("You don't have sufficient balance")
else:
    print("Your account is inactive")
```

```python
#Create a restaurant billing system that calculates a discount based on the total bill and customer membership.

cust_membership = input("Do the customer have the membership?(y/n)")

if cust_membership == "y":
    bill = int(input("Enter customer's billing amount: "))
    if bill >= 1000 and bill<2000:
        print("Customer get basic membership discount of '10%' and discount:", bill*0.1)
    elif bill >= 2000 and bill<5000:
        print("Customer get prime membership discount of '20%' and discount:", bill*0.2)
    elif bill >= 5000 and bill<10000:
        print("Customer get pro membership discount of '30%' and discount:", bill*0.3)
    elif bill >= 10000:
        print("Customer get max membership discount of '50%' and discount:", bill*0.5)
    else:
        print("Invalid input")
else:
    print("No discount")
```
