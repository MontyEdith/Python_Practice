```python
#Find the largest among four numbers using only conditional statements.

num1 = float(input("Enter 1st number: "))
num2 = float(input("Enter 2nd number: "))
num3 = float(input("Enter 3rd number: "))
num4 = float(input("Enter 4th number: "))

print(num1,num2,num3,num4)

if num1>num2:
    if num1>num3:
        if num1>num4:
            print("Largest number: ", num1)
        else:
            print("Largest number: ", num4)
    else:
        if num3>num4:
            print("Largest number: ", num3)
        else:
            print("Largest number: ", num4)
else:
    if num2>num3:
        if num2>num4:
            print("Largest number: ", num2)
        else:
            print("Largest number: ", num4)
    else:
        if num3>num4:
            print("Largest number: ", num3)
        else:
            print("Largest number: ", num4)
```

```python
#Find the second-largest among three numbers using if-elif-else.

num1 = float(input("Enter 1st number: "))
num2 = float(input("Enter 2nd number: "))
num3 = float(input("Enter 3rd number: "))

print(num1,num2,num3)

if num1 > num2 and num1 > num3:                                 #num1 is largest
    if num2 > num3:
        print("Second-largest among three numbers: ", num2)
    else:
        print("Second-largest among three numbers: ", num3)

elif num2 > num1 and num2 > num3:                               #num2 is largest
    if num1 > num3:
        print("Second-largest among three numbers: ", num1)
    else:
        print("Second-largest among three numbers: ", num3)

else:                                                            #num1 is largest
    if num1 > num2:
        print("Second-largest among three numbers: ", num1)
    else:
        print("Second-largest among three numbers: ", num2)
```

```python
"""Check whether a given year is a leap year using the complete leap-year rule:
Divisible by 400, or
Divisible by 4 but not by 100."""

year = int(input("Enter your year: "))

if (year%400==0) or (year%100!=0 and year%4==0):
    print("Leap year")
else:
    print("Not a Leap year")
```

```python
"""Given three subject marks, determine:
Whether the student passed every subject
Total marks
Percentage
Grade"""

math = float(input("Enter your mathematics marks: "))
physics = float(input("Enter your physics marks: "))
chemistry = float(input("Enter your chemistry marks: "))

print(f"Student scored {math} in mathematics")
print(f"Student scored {physics} in physics")
print(f"Student scored {chemistry} in chemistry\n")


if math >= 40 and physics >= 40:
    if chemistry >=40:
        print("Student passed every subject")
    else:
        print("Student passed math and physics but failed in chemistry")
elif chemistry >= 40 and physics >= 40:
    if math >=40:
        print("Student passed every subject")
    else:
        print("Student passed chemistry and physics but failed in math")
elif math >= 40 and chemistry >= 40:
    if physics >=40:
        print("Student passed every subject")
    else:
        print("Student passed math and chemistry but failed in physics")
else:
    print("Invalid Input")

total = math+physics+chemistry
print(f"\nStudents total marks out of 300 is {total}\n")

percent = (total/300)*100
print(f"Student's percentage as per the total marks is {percent}\n")

if percent>=90 and percent<=100:
    print("Student scored grade A")
elif percent>=75 and percent<90:
    print("Student scored grade B")
elif percent>=60 and percent<75:
    print("Student scored grade C")
elif percent>=40 and percent<60:
    print("Student scored grade D")
elif percent>=0 and percent<40:
    print("Student scored grade E")
else:
    print("Invalid Input")
```

```python
"""Create a password-strength checker using conditions:
At least 8 characters
Contains a digit
Contains an uppercase letter"""

password = input("Enter your password: ")

digit = False
upper = False

for char in password:
    if char.isdigit():
        digit = True
    elif char.isupper():
        upper = True
    else:
        pass

if digit == True and upper == True and len(password) >= 8:
    print("Strong Password")
else:
    print("Weak Password")
```

```python
#Create a simple rock-paper-scissors program using conditional statements.

import random

user = input("Enter your choice:(rock/paper/scissors)")

computer = random.choice(["rock","paper","scissors"])

print(f"Computer chose {computer}\n")

if user == computer:
    print("It's a tie")

elif user == "rock":
    if computer == "paper":
        print("Computer wins")
    else:
        print("User wins")

elif user == "paper":
    if computer == "scissors":
        print("Computer wins")
    else:
        print("User wins")

elif user == "scissors":
    if computer == "rock":
        print("Computer wins")
    else:
        print("User wins")

else:
    print("Invalid Input")
```

```python
#Create a calculator that handles division by zero using conditional statements.

num1 = float(input("Enter 1st number: "))
num2 = float(input("Enter 2nd number: "))

print("Enter '+' for Addition")
print("Enter '-' for Subtraction")
print("Enter '*' for Multiplication")
print("Enter '/' for Division")
oper = input("Enter your operation choice(+,-,*,/)")

if oper == "+":
    print("Addition:",num1+num2)
elif oper == "-":
    print("Subtraction:",num1-num2)
elif oper == "*":
    print("Multiplication:",num1*num2)
elif oper == "/":
    if num2 == 0:
        print("Can't do zero division")
    else:
        print("Division:",num1/num2)
else:
    print("Invalid Input")
```
