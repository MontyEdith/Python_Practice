```python
#Find the sum of all even numbers from 1 to 100.

sum = 0

for num in range(1,101):
    if num%2 == 0:
        sum += num

print(f"Sum of all even numbers from 1 to 100 is {sum}")
```

```python
#Count how many numbers between 1 and 100 are divisible by 3.

count = 0

for num in range(1,101):
    if num%3 == 0:
        count += 1

print(f"Numbers between 1 and 100 are divisible by 3 are {count}")
```

```python
#Take 10 numbers from the user and count how many are positive.

count = 0
for i in range(10):
    num = float(input(f"Enter number {i+1}: "))
    if num >0:
        count += 1

print(f"Out of 10 numbers entered by user, {count} are positive")
```

```python
#Take 10 numbers from the user and count how many are negative.

count = 0
for i in range(10):
    num = float(input(f"Enter number {i+1}: "))
    if num < 0:
        count += 1

print(f"Out of 10 numbers entered by user, {count} are negative")
```

```python
#Take 10 numbers from the user and count how many are zero.

count = 0
for i in range(10):
    num = float(input(f"Enter number {i+1}: "))
    if num == 0:
        count += 1

print(f"Out of 10 numbers entered by user, {count} are zero")
```

```python
#Take 10 numbers and count how many are even and how many are odd.

even_count = 0
odd_count = 0
for i in range(10):
    num = float(input(f"Enter number {i+1}: "))
    print(f"Entered number {i+1}: {num} ")
    if num%2 == 0:
        even_count += 1
    else:
        odd_count += 1

print(f"Out of 10 numbers entered by user, {even_count} are even and {odd_count} are odd")
```

```python
#Take n numbers from the user and find the largest number.

n = int(input("How many number you want to enter: "))
l1 = []
for i in range(n):
    num = int(input(f"Enter number {i+1}: "))
    print(f"Entered number {i+1}: {num} ")
    l1.append(num)

largest = l1[0]
for i in l1:
    if i > largest:
        largest = i

print(f"Out of {n} numbers entered by user, largest number is {largest}")
```

```python
#Take n numbers from the user and find the smallest number.

n = int(input("How many number you want to enter: "))
l1 = []
for i in range(n):
    num = int(input(f"Enter number {i+1}: "))
    print(f"Entered number {i+1}: {num} ")
    l1.append(num)

smallest = l1[0]
for i in l1:
    if i < smallest:
        smallest = i

print(f"Out of {n} numbers entered by user, smallest number is {smallest}")
```

```python
#Take 10 numbers and find the sum of only the positive numbers.

sum = 0
count = 0
for i in range(10):
    num = int(input(f"Enter number {i+1}: "))
    print(f"Entered number {i+1}: {num} ")
    if num > 0:
        sum += num
        count += 1

print(f"Out of 10 numbers entered by user, {count} are positive and their sum is {sum}")
```

```python
#Take 10 numbers and find the sum of only the even numbers.

sum = 0
count = 0
for i in range(10):
    num = int(input(f"Enter number {i+1}: "))
    print(f"Entered number {i+1}: {num} ")
    if num%2 == 0:
        sum += num
        count += 1

print(f"Out of 10 numbers entered by user, {count} are even and their sum is {sum}")
```
