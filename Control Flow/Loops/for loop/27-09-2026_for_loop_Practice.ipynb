```python
"""Calculate the factorial of a number.

Example:

Input: 5
Output: 120"""

num = int(input("Enter any number: "))

fact = 1
for i in range(1,num+1):
    fact *=i

print(f"The factorial of {num} is {fact}")
```

```python
#Calculate x^n using a for loop without using **.

num = int(input("Enter any number: "))
power = int(input("Enter the power: "))

for i in range(1,power+1):
    if i == 1:
        result = num
    else:
        result *=num

print(f"The {num} to the power of {power} is {result}")
```

```python
#Find the sum of squares from 1 to n.

n = int(input("Enter any number: "))

sum = 0
for i in range(1,n+1):
    sum += i**2

print(f"The sum of squares from 1 to {n} is {sum}")
```

```python
#Find the sum of cubes from 1 to n.

n = int(input("Enter any number: "))

sum = 0
for i in range(1,n+1):
    sum += i**3

print(f"The sum of cubes from 1 to {n} is {sum}")
```

```python
#Print the first 10 multiples of a given number.

n = int(input("Enter any number: \n"))

print(f"The first 10 multiples of a given {n} is: \n")
mul = 1
for i in range(1,11):
    mul = n * i
    print(f"The {i} multiple of {n} is {mul}")
```

```python
#Calculate the product of numbers from 1 to n.

n = int(input("Enter any number: \n"))
prod = 1
for i in range(1,n+1):
    prod *= i

print(f"\nThe product of numbers from 1 to {n} is {prod}")
```

```python
#Find all factors of a given number.

n = int(input("Enter any number: \n"))
l = []
for i in range(1,n+1):
    if n%i == 0:
        l.append(i)

print(f"All factors of {n} are {l}")
```

```python
#Count the number of factors of a given number.

n = int(input("Enter any number: \n"))
count = 0
for i in range(1,n+1):
    if n%i == 0:
        count +=1

print(f"The number of factors of {n} are {count}")
```

```python
#Check whether a number is prime.

n = int(input("Enter any number: \n"))

prime = True

for i in range(2,n):
    if n%i == 0:
        prime = False
        break

if prime:
    print(f"{n} is a prime number")
else:
    print(f"{n} is not a prime number")
```

```python
#Print all prime numbers between 1 and 100.

l = []
for i in range(2,101):
    prime = True
    for j in range(2,i):
        if i%j == 0:
            prime = False
            break
    if prime:
        l.append(j)

print(f"All prime numbers between 1 and 100 are {[l]}")
print(len(l))

```

```python
#Print all prime numbers between two numbers entered by the user.

l = []
num1 = int(input("Enter 1st number: \n"))
num2 = int(input("Enter 2nd number: \n"))

for i in range(num1,num2+1):
    prime = True
    for j in range(2,i):
        if i%j == 0:
            prime = False
            break
    if prime:
        l.append(i)

print(f"All prime numbers between {num1} and {num2} are {l}")
```

```python
#Count the number of prime numbers between 1 and 100.

count = 0

for i in range(2,101):
    prime = True
    for j in range(2,i):
        if i%j == 0:
            prime = False
            break
    if prime:
        count += 1

print(f"The number of prime numbers between 1 and 100 are {count}")
```

```python
#Find the sum of all prime numbers between 1 and 100.

sum = 0

for i in range(2,101):
    prime = True
    for j in range(2,i):
        if i%j == 0:
            prime = False
            break
    if prime:
        sum += i

print(f"The sum of all prime numbers between 1 and 100 are {sum}")
```

```python
#Check whether a number is a perfect number.

n = int(input("Enter any number: \n"))
sum = 0

for i in range(1,n):
    if n%i == 0:
        sum +=i

if sum == n:
    print(f"{n} is a perfect number")
else:
    print(f"{n} is not a perfect number")
```

```python
#Print all perfect numbers between 1 and 1000.

perfect_num = []
for n in range(1,1001):
    sum=0
    for i in range(1,n):
        if n%i == 0:
            sum +=i

    if sum == n:
        perfect_num.append(n)

print(f"All perfect numbers between 1 and 1000 are {perfect_num}")
```

```python
#Check whether a number is an Armstrong number.
#1³ + 5³ + 3³ = 1 + 125 + 27 = 153

num = input("Enter any number: \n")

power = len(num)
sum = 0

for i in num:
    sum +=int(i)**power

if sum == int(num):
    print(f"{num} is Armstrong number")
else:
    print(f"{num} is not an Armstrong number")
```

```python
#Print all Armstrong numbers between 1 and 1000.

arms_num = []
for num in range(1,1001):
    power = len(str(num))
    sum = 0
    for i in str(num):
        sum +=int(i)**power

    if sum == int(num):
        arms_num.append(num)
    else:
        pass

print(f"All Armstrong numbers between 1 and 1000 are {arms_num}")
```

```python
#Find the GCD/HCF of two numbers using a for loop.

num1 = int(input("Enter 1st number: "))
num2 = int(input("Enter 2nd number: "))

hcf = 1
for i in range(1,min(num1,num2)+1):
    if num1%i == 0 and num2%i == 0:
        hcf = i

print(f"The GCD/HCF of {num1} and {num2} is {hcf}")

```

```python
#Count the number of characters in a string without using len().

user = input("Enter any string: ")
count = 0
for i in user:
    count +=1

print(f"The number of characters in string {user} is {count}")
```

```python
#Count the number of vowels in a string.

vowel = ["a","e","i","o","u"]

user = input("Enter any string: ")
count = 0
for i in user:
    if i in vowel:
        count +=1

print(f"The number of vowels in string {user} is {count}")

```

```python
#Count the number of consonants in a string.

vowel = ["a","e","i","o","u"]

user = input("Enter any string: ")
count = 0
for i in user:
    if i not in vowel:
        count +=1
        

print(f"The number of consonants in string {user} is {count}")
```

```python
#Count the number of digits in a string.

user = input("Enter any string: ")
count = 0
for i in user:
    if i.isdigit():
        count +=1
        

print(f"The number of digits in string {user} is {count}")
```

```python
#Count uppercase letters in a string.

user = input("Enter any string: ")
count = 0
for i in user:
    if i == i.upper():
        count +=1
        

print(f"The number of uppercase letters in string {user} is {count}")
```

```python
#Count lowercase letters in a string.

user = input("Enter any string: ")
count = 0
for i in user:
    if i == i.lower():
        count +=1
        

print(f"The number of lowercase letters in string {user} is {count}")
```

```python
#Count spaces in a string.

user = input("Enter any string: ")
count = 0
for i in user:
    if i == " ":
        count +=1
        

print(f"The number of spaces in string {user} is {count}")
```
