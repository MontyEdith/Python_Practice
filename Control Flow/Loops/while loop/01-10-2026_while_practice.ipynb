```python
#28. Calculate a number raised to a power without using the ** operator.

num = int(input("Enter any number: "))
power = int(input("Enter power: "))

result = 1
count = 1

while count <= power:
    result *= num
    count += 1

print(result)
```

```python
#29. Find the sum of all numbers divisible by 3 or 5 between 1 and n.

num = 1
n = int(input("Enter any number: "))
sum = 0
div = []
while num < (n + 1):
    if num%3 == 0 or num%5 == 0:
        div.append(num)
        sum += num
    num += 1

print(div)
print(sum)
```

```python
#31. Count the number of digits in an integer.

num = int(input("Enter any number: "))
count = 0
n = num

if num == 0:
    count = 1
else:
    num = abs(num)

while num > 0:
    num = num//10
    count += 1

print(f"Number of digits in {n} is {count}")
```

```python
#32. Calculate the sum of the digits of a number.

num = int(input("Enter any number: "))
n = num
sum = 0

if num == 0:
    sum = 0
else:
    num = abs(num)

while num > 0:
    a = num%10
    num = num//10
    sum += a

print(f"The sum of the digits of {n} is {sum}")
```

```python
#33. Calculate the product of the digits of a number.

num = int(input("Enter any number: "))
n = num
prod = 1

if num == 0:
    prod = 0
else:
    num = abs(num)

while num > 0:
    a = num%10
    print(a)
    num = num//10
    prod *= a

print(f"The product of the digits of {n} is {prod}")
```

```python
#34. Reverse a given integer.

num = int(input("Enter any number: "))
n = num
rev = 0

if num == 0:
    rev = 0
else:
    num = abs(num)

while num > 0:
    a = num%10
    num = num//10
    rev = rev*10 + a

print(f"The product of the digits of {n} is {rev}")
```

```python
#35. Check whether a number is a palindrome.

num = int(input("Enter any number: "))
n = num
rev = 0

while num > 0:
    a = num%10
    num = num//10
    rev = rev*10 + a

if n == rev:
    print(f"{n} is a palindrome")
else:
    print(f"{n} is not a palindrome")
```

```python
#36. Find the largest digit in a number.

num = int(input("Enter any number: "))
n = num
largest = 0

while num > 0:
    a = num%10
    num = num//10
    if a > largest:
        largest = a

print(f"Largest digit in {n} is {largest}")
```

```python
#37. Find the smallest digit in a number.

num = int(input("Enter any number: "))
n = num
smallest = 9

while num > 0:
    a = num%10
    num = num//10
    if a < smallest:
        smallest = a

print(f"Largest digit in {n} is {smallest}")
```

```python
#38. Count how many even and odd digits are in a number.

num = int(input("Enter any number: "))
n = num
even_count = 0
odd_count = 0

while num > 0:
    a = num%10
    num = num//10
    if a%2 == 0:
        even_count += 1
    elif a%2 != 0:
        odd_count += 1

print(f"Total {even_count} even numbers and {odd_count} odd numbers are there in {n}")
```

```python
#39. Count the occurrences of a particular digit in a number.

num = int(input("Enter any number: "))
digit = int(input("Enter the digit you want the occurrence: "))
n = num
count = 0

while num > 0:
    a = num%10
    num = num//10
    if a == digit:
        count += 1

print(f"{digit} occurred {count} times in {n}")
```

```python
#40. Check whether a number contains the digit 0.

num = int(input("Enter any number: "))
n = num
count = 0

while num > 0:
    a = num%10
    num = num//10
    if a == 0:
        count = 1

if count == 1:
    print(f"{n} contains the digit 0")
else:
    print(f"{n} does not contains the digit 0")
```

```python
#43. Check whether a number is an Armstrong number.

num = int(input("Enter any number: "))
n = num
count = 0
arm = 0

while num > 0:
    count += 1
    num = num//10

num = n

while num > 0:
    arm = arm + ((num%10) ** count)
    num = num//10

if arm == n:
    print(f"{n} is an Armstrong number")
else:
    print(f"{n} is not an Armstrong number")
```

```python
#44. Check whether a number is a strong number.

num = int(input("Enter any number: "))
n = num
total = 0

while num > 0:
    a = num%10

    fact = 1
    i = 1

    while i <= a:
        fact *= i
        i += 1

    total +=fact
    num //= 10

if total == n:
    print(f"{n} is a strong number")
else:
    print(f"{n} is not a strong number")
```

```python
#45. Find the sum of the digits until the result becomes a single digit.

num = int(input("Enter any number: "))
n = num

while num >=10:
    total = 0

    while num > 0:
        a = num%10
        total += a
        num //= 10

    num = total

print(f"Single digit sum: {total}")
```
