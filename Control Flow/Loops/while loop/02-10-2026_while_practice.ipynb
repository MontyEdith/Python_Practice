```python
#46. Check whether a number is prime.

num = int(input("Enter any number: "))
n = 2
prime = True

while n < num:
    if num%n == 0:
        prime = False
        break
    n += 1

if prime:
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
```

```python
#47. Print all prime numbers between 1 and n.

n = int(input("Enter any number: "))

num = 2
prime_num = []

while num <= n:
    a = 2
    prime = True
    while a < num:
        if num%a == 0:
            prime = False
            break
        a += 1
    if prime:
        prime_num.append(num)
    num +=1


print(f"All prime numbers between 1 and {n} are {prime_num}")
```

```python
#51. Calculate the HCF/GCD of two numbers.

num1 = int(input("Enter any number:"))
num2 = int(input("Enter any number:"))

factors_num1 = []
factors_num2 = []

n = 1
while n < num1:
    if num1%n == 0:
        factors_num1.append(n)
    n += 1

n = 1
while n < num2:
    if num2%n == 0:
        factors_num2.append(n)
    n += 1

common_factors = []

n = 0

while n < len(factors_num1):
    if factors_num1[n] in factors_num2:
        common_factors.append(factors_num1[n])
    n += 1

print(f"HCF/GCD of {num1} and {num2} is {common_factors[-1]}")
```

```python
#52. Calculate the LCM of two numbers.

num1 = int(input("Enter 1 st number: "))
num2 = int(input("Enter 1 st number: "))

num1_multiples = []
num2_multiples = []
n = 1

while n < num2:
    num1_multiples.append(n*num1)
    n += 1

n = 1

while n < num1:
    num2_multiples.append(n*num2)
    n += 1


common_multiples = []

n = 0
while n < len(num1_multiples):
    if num1_multiples[n] in num2_multiples:
        common_multiples.append(num1_multiples[n])
    n += 1

print(f"{num1_multiples}")
print(f"{num2_multiples}")
print(f"{common_multiples}")
print(f"{common_multiples[0]}")
```

```python
#53. Check whether a number is a perfect number.

num = int(input("Enter any number: "))
n = 1
total = 0

while n < num:
    if num%n == 0:
        total += n
    n += 1

if total == num:
    print(f"{num} is a perfect number")
else:
    print(f"{num} is not a perfect number")
```

```python
#55. Check whether a number is a Harshad number.

#1729: The sum of the digits is \(1 + 7 + 2 + 9 = 19\). Because 1729 is divisible by 19 (\(1729 \div 19 = 91\)), it is a Harshad number.

num = int(input("Enter any number: "))
n = num
total = 0

while num > 0:
    digit = num%10
    total += digit
    num //= 10
    

if n%total == 0:
    print(f"{n} is a Harshad number")
else:
    print(f"{n} is not a Harshad number")
```

```python
#56. Check whether a number is a neon number.

# 81 = 9*9 and 81 => 8+1 = 9

num = int(input("Enter any number: "))
n = num
total = 0

while num > 0:
    digit = num%10
    total += digit
    num //= 10

if (n == total**2):
    print(f"{n} is a neon number")
else:
    print(f"{n} is not a neon number")
```

```python
#107. Find the nth Fibonacci number using a while loop.

n = int(input("Enter index for which you want to find Fibonacci number?"))
a = 0
b = 1
count = 0

while count < n + 1:
    print(a,end=" ")
    if count == n:
        break
    else:
        next_term = a+b
        a = b
        b = next_term

    count += 1

print(f"The {n} Fibonacci number is {a}")
```
