```python
# 13. Create a Pandas Line Plot
# Create a Pandas DataFrame containing monthly revenue and use Pandas' built-in .plot() method to visualize it.

import pandas as pd

months = ["January","February","March","April","May","June","July","August","September","October","November","December"]
monthly_sales = [123,546,321,457,987,567,897,321,765,543,897,234]
quantity = [4,9,7,6,5,2,4,5,7,1,2,9]

df = pd.DataFrame({
    "Months" : months,
    "Monthly Sales" : monthly_sales,
    "Quantity" : quantity
})

df["Revenue"] = df["Monthly Sales"] * df["Quantity"]

df.plot(
    x="Months",
    y="Revenue",
    kind="line",
    figsize=(8,6),
    marker="s",
    title="Monthly Revenue"
)

plt.ylabel("Revenue")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 17. Create a Stacked Bar Chart
# Create a stacked bar chart showing sales of two categories across multiple products.

import pandas as pd

df = pd.DataFrame({
    "Product": [
        "Laptop", "Laptop",
        "Mobile", "Mobile",
        "Tablet", "Tablet",
        "Headphones", "Headphones"
    ],
    "Category": [
        "Online", "Retail",
        "Online", "Retail",
        "Online", "Retail",
        "Online", "Retail"
    ],
    "Sales": [
        30000, 15000,
        20000, 10000,
        12000, 6000,
        3000, 2000
    ]
})

pivot_df = df.pivot(
    index="Product",
    columns="Category",
    values="Sales"
)

plt.figure(figsize=(8, 6))

pivot_df.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 6)
)

plt.title("Product-wise Sales by Category")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.legend(title="Category")
plt.tight_layout()
plt.show()
```

```python
# 26. Create a 2×2 Subplot Grid
# Create four different plots in a single figure:
# - Line plot
# - Bar plot
# - Scatter plot
# - Histogram

df = pd.DataFrame({
    "Product": [
        "Laptop", "Mobile", "Tablet", "Headphones",
        "Smartwatch", "Keyboard", "Mouse", "Monitor",
        "Printer", "Speaker"
    ],
    "Sales": [
        45000, 72000, 28000, 15000, 35000,
        12000, 8000, 55000, 22000, 18000
    ]
})

fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))
axs[0,0].plot(df["Sales"],marker='s', color='green', label='Sales')
axs[0,0].set_title("Product sales")
axs[0,0].set_xlabel("Product")
axs[0,0].set_ylabel("Sales")
axs[0,0].legend()
axs[0,0].grid(True)

axs[0,1].bar(df["Product"],df["Sales"])
axs[0,1].set_title("Product sales")
axs[0,1].set_xlabel("Product")
axs[0,1].tick_params(axis="x", labelrotation=45)
axs[0,1].set_ylabel("Sales")
axs[0,1].grid(True)

axs[1,0].scatter(df["Product"],df["Sales"])
axs[1,0].set_title("Product sales")
axs[1,0].set_xlabel("Product")
axs[1,0].tick_params(axis="x", labelrotation=45)
axs[1,0].set_ylabel("Sales")
axs[1,0].grid(True)

axs[1,1].hist(df["Sales"], bins=10)
axs[1,1].set_title("Product sales")
axs[1,1].set_xlabel("Sales")
axs[1,1].set_ylabel("Frequency")
axs[1,1].tick_params(axis="x", labelrotation=45)
axs[1,1].set_ylabel("Sales")
axs[1,1].grid(True)

plt.tight_layout()
plt.show()
```

```python
# 29. Monthly Revenue Analysis
# Create a dataset containing:
# - Date
# - Product
# - Quantity
# - Price
# Calculate:
# Revenue = Quantity * Price


# Then visualize monthly revenue using a line plot.

df = pd.DataFrame({
    "Date": [
        "2026-07-01", "2026-11-02", "2026-01-03", "2026-01-04",
        "2026-02-05", "2026-12-06", "2026-08-07", "2026-07-08",
        "2026-05-09", "2026-02-10", "2026-11-11", "2026-06-12",
        "2026-03-13", "2026-02-14", "2026-10-15", "2026-09-16",
        "2026-09-17", "2026-07-18", "2026-05-19", "2026-04-20"
    ],
    "Product": [
        "Laptop", "Mobile", "Tablet", "Headphones", "Keyboard",
        "Mouse", "Monitor", "Laptop", "Mobile", "Speaker",
        "Tablet", "Headphones", "Printer", "Keyboard", "Mobile",
        "Laptop", "Mouse", "Monitor", "Speaker", "Tablet"
    ],
    "Quantity": [
        2, 5, 3, 4, 6, 10, 2, 1, 3, 5,
        4, 7, 2, 8, 6, 3, 12, 1, 4, 5
    ],
    "Price": [
        55000, 20000, 30000, 2500, 1200,
        600, 18000, 52000, 22000, 3500,
        28000, 3000, 12000, 1500, 21000,
        58000, 650, 19000, 4000, 32000
    ]
})

df["Revenue"] = df["Quantity"] * df["Price"]
df["Date"] = pd.to_datetime(df["Date"])
monthly_revenue = df.groupby(df["Date"].dt.to_period("M"))["Revenue"].sum()

plt.plot(monthly_revenue.index.astype(str), monthly_revenue.values, marker="o")
plt.xlabel("Month")
plt.xticks(rotation=45)
plt.ylabel("Revenue")
plt.title("Monthly Revenue Analysis")
plt.show()
```
