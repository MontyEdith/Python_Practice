```python
"""Word Frequency Analyzer

Ask the user for a sentence and create a dictionary containing the frequency of every word.

Example:

Input:
python is easy and python is powerful

Output:

{
    "python": 2,
    "is": 2,
    "easy": 1,
    "and": 1,
    "powerful": 1
}"""

user_input = input("Enter any sentence: ")

words = list(user_input.split())

frequency = dict()

for i in words:
    frequency[i] = frequency.get(i,0) + 1

print(frequency)


```

```python
"""Bank Account System

Create a dictionary containing account numbers and balances.

Example:

accounts = {
    101: 50000,
    102: 25000,
    103: 75000
}

Allow the user to:

Check balance
Deposit money
Withdraw money
Create account
Delete account"""

accounts = {
    101: 50000,
    102: 25000,
    103: 75000
}

print("Bank Account System")
print("Enter 1 for Check balance")
print("Enter 2 for Deposit money")
print("Enter 3 for Withdraw money")
print("Enter 4 for Create account")
print("Enter 5 for Delete account")
user_choice = int(input("Enter your choice: "))

if user_choice == 1:
    account_num = int(input("Enter your account number: "))
    for k,v in accounts.items():
        if account_num == k:
            print(f"Your balance is {v}")

elif user_choice == 2:
    account_num = int(input("Enter your account number: "))
    depo_amt = int(input("Enter the amount you want to deposit: "))
    for k,v in accounts.items():
        if account_num == k:
            v += depo_amt
            accounts[account_num] = v
            print(f"Your balance after deposit is {v}")

elif user_choice == 3:
    account_num = int(input("Enter your account number: "))
    withdraw_amt = int(input("Enter the amount you want to withdraw: "))
    for k,v in accounts.items():
        if account_num == k:
            if withdraw_amt <= v:
                v -= withdraw_amt
                accounts[account_num] = v
                print(f"Your balance after withdraw is {v}")
            else:
                print("You don't have enough balance")

elif user_choice == 4:
    account_num = int(input("Enter your account number: "))
    amt = int(input("Enter the balance you have: "))
    accounts[account_num] = amt
    print("Account created")

elif user_choice == 5:
    account_num = int(input("Enter account number you want to delete: "))
    del accounts[account_num]
    print("Account deleted")

accounts_display = input("Do you want see all accounts? (y/n)")

if accounts_display == "y":
    print(accounts)
elif accounts_display == "n":
    print("Okay, Have a good day")



```
