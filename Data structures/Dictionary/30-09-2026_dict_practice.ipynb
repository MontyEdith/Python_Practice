```python
"""Count the frequency of each character in a string.
"hello"

Expected:

{'h': 1, 'e': 1, 'l': 2, 'o': 1}"""

string = "machinelearningengineer"

frequency = {}

for char in string:
    frequency[char] = frequency.get(char, 0) + 1

print(frequency)
```

```python
#Count the frequency of each word in a sentence.

string = "the dog saw the cat and the dog ran"

frequency = {}

for i in string.split():
    frequency[i] = frequency.get(i,0) + 1

print(frequency)
```

```python
#Find the most frequently occurring character in a string.

string = "machinelearningengineer"

frequency = dict()

for i in string:
    frequency[i] = frequency.get(i,0) + 1

a = list(frequency.values())[0]
b = list(frequency.keys())[0]

for k,v in frequency.items():
    if v > a:
        a = v
        b = k

print(f"The most frequently occurring character in a string is {b}")
```

```python
#Count vowels and consonants using a dictionary.

string = "machinelearningengineer"

frequency = dict()

for i in string:
    frequency[i] = frequency.get(i,0) + 1

a = ["a","e","i","o","u"]
vowel_count = 0
consonants_count = 0

for k,v in frequency.items():
    if k in a:
        vowel_count += 1
    else:
        consonants_count += 1

print(f"Vowel count: {vowel_count}")
print(f"Consonants count: {consonants_count}")
```

```python
#Count uppercase, lowercase, digits, and special characters in a string using a dictionary.

string = "Machine1Learning2Engineer4*^*"

frequency = dict()

for i in string:
    frequency[i] = frequency.get(i,0) + 1

uppercase_count = 0
lowercase_count = 0
digits_count = 0
special_characters_count = 0

for k,v in frequency.items():
    if k.isupper():
        uppercase_count += v
    elif k.islower():
        lowercase_count += v
    elif k.isdigit():
        digits_count += v
    else:
        special_characters_count += v

print(f"Uppercase count: {uppercase_count}")
print(f"Lowercase count: {lowercase_count}")
print(f"Digits count: {digits_count}")
print(f"Special characters count: {special_characters_count}")
```
