```python
"""Print Even Numbers
Given:

numbers = [12, 5, 8, 21, 30, 17, 44]

Create a new list containing only the even numbers."""

numbers = [12, 5, 8, 21, 30, 17, 44]

even_num = []

for i in numbers:
    if i%2 == 0:
        even_num.append(i)

print(f"Even number list: {even_num}")

```

```python
"""Print Odd Numbers
Given a list of numbers, create a new list containing only the odd numbers."""

numbers = [12, 5, 8, 21, 30, 17, 44]

odd_num = []

for i in numbers:
    if i%2 != 0:
        odd_num.append(i)

print(f"Odd number list: {odd_num}")
```

```python
"""Positive and Negative Numbers
Given:

numbers = [-10, 20, -5, 30, 0, -15, 40]

Create separate lists for positive and negative numbers."""

numbers = [-10, 20, -5, 30, 0, -15, 40]

pos_num = []
neg_num = []

for i in numbers:
    if i > 0:
        pos_num.append(i)
    elif i < 0:
        neg_num.append(i)
    else:
        pass

print(f"Positive number list: {pos_num}")
print(f"Negative number list: {neg_num}")
```

```python
"""Remove Duplicates
Given:

numbers = [10, 20, 10, 30, 20, 40, 10, 50]

Create a new list containing each number only once."""

numbers = [10, 20, 10, 30, 20, 40, 10, 50]
new_num = []

for i in set(numbers):
    new_num.append(i)
new_num.sort()

print(f"New list without duplicates: {new_num}")
```

```python
"""Find Second Largest Number
Given a list of numbers, find the second-largest number without directly using sorted()."""

numbers = [10, 20, 10, 30, 20, 40, 10, 50]

largest = numbers[0]
second_largest = numbers[0]

for i in numbers:
    if i > largest:
        second_largest = largest
        largest = i

    elif i > second_largest and i != largest:
        second_largest =  i

print(f"Second Largest Number: {second_largest}")


```

```python
"""Replace Negative Numbers
Given:

numbers = [10, -5, 20, -8, 30, -2]

Replace every negative number with 0."""

numbers = [10, -5, 20, -8, 30, -2]

for i in range(len(numbers)):
    if numbers[i] < 0:
        numbers[i] = 0

print(numbers)

```

```python
"""Remove All Occurrences
Given:

numbers = [10, 20, 10, 30, 10, 40, 10]

Remove all occurrences of 10."""

numbers = [10, 20, 10, 30, 10, 40, 10]
new_numbers = []

for i in numbers:
    if i != 10:
        new_numbers.append(i)

print(new_numbers)
```

```python
"""Find Common Elements
Given:

list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

Create a new list containing the common elements."""

list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
new_list = []

for i in list1:
    if i in list2:
        new_list.append(i)

print(new_list)
```

```python
"""Separate Even and Odd Numbers
Ask the user to enter n numbers. Store them in a list and then create:
even_numbers
odd_numbers"""

import random

n = random.choice(range(1,10))
even_num = []
odd_num = []
num_list = []

for i in range(n):
    num = int(input("Enter any number: "))
    if num%2 == 0:
        even_num.append(num)
    else:
        odd_num.append(num)
    num_list.append(num)

print(num_list)
print(even_num)
print(odd_num)
```

```python
"""📊 List Analysis Challenge
Ask the user how many numbers they want to enter. Store all numbers in a list and display:
Total number of elements
Sum
Average
Largest number
Smallest number
Number of even numbers
Number of odd numbers
Number of positive numbers
Number of negative numbers
Number of zeros
List in ascending order
List in descending order
List without duplicates"""

n = int(input("How many numbers you want to enter: "))
even_count = 0
odd_count = 0
num_list = []
pos_count = 0
neg_count = 0
zero_count = 0

for i in range(n):
    num = int(input("Enter any number: "))
    num_list.append(num)
    if num%2 == 0:
        even_count += 1
    else:
        odd_count += 1
    
    if num > 0:
        pos_count += 1
    elif num < 0:
        neg_count += 1
    else:
        zero_count += 1


print(num_list)
print(f"Total number of elements are {len(num_list)}")
print(f"Sum: {sum(num_list)}")
print(f"Average: {sum(num_list)/len(num_list)}")
print(f"Largest number: {max(num_list)}")
print(f"Smallest number: {min(num_list)}")
print(f"Number of even numbers: {even_count}")
print(f"Number of odd numbers: {odd_count}")
print(f"Number of positive numbers: {pos_count}")
print(f"Number of negative numbers: {neg_count}")
print(f"Number of zeros: {zero_count}")
num_list.sort()
print(f"List in ascending order: {num_list}")
num_list.sort(reverse=True)
print(f"List in descending order: {num_list}")
num_list = set(num_list)
num_list = list(num_list)
print(f"List without duplicates: {num_list}")
```
