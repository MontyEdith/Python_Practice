```python
# [ ] 1. Write a program that takes two numbers from the user and handles a ZeroDivisionError when dividing them.

num1 = int(input("Enter any number: "))
num2 = int(input("Enter any number: "))


def divide():
    try:
        final = num1/num2
        return final
    
    except ZeroDivisionError:
        print("Cannot do division by zero")


result = divide()

if result is not None:
    print(f"Result: {result:.2f}")
```

```python
# [ ] 9. Create a list of numbers and attempt to divide each number by a user-provided divisor. Handle division by zero.

numbers = [98,76,54,32,10]

def divide():
    new_numbers = []
    
    try:
        n = int(input("Enter any number: "))
        for i in numbers:
            new_numbers.append(i/n)
        print(new_numbers)
    except ZeroDivisionError:
        print("Error: Can't divide by zero")
    except ValueError:
        print("Error: Please enter valid integers")

divide()
```

```python
# [ ] 16. Create a program that opens a file, reads its contents, and uses finally to close the file. Then rewrite it using with open().

import os

def read_text_file():
    filename = "kholo.txt"
    filepath = os.path.join(os.getcwd(), filename)

    try:
        f = open(filepath, "r")

    except FileNotFoundError:
        print("Error: No such file exist in current directory")

    except PermissionError:
        print("Error: You don't have permission to read the file")

    else:
        print(f.read())

    finally:
        if f is not None:
            f.close()
    

read_text_file()
```

```python
# [ ] 17. Ask the user for their age. Handle non-numeric input and raise an error if the age is negative.

class AgeNotNegative(Exception):
    pass

def user_age():
    try:
        age = int(input("Enter your age:"))
        if age < 0:
            raise AgeNotNegative("Age can't be zero")

        print(f"Age: {age}")

    except ValueError:
        print("Error: Please enter valid age")

    except AgeNotNegative as e:
        print(f"Error: {e}")

user_age()
```

```python
# [ ] 19. Create a calculator using functions for addition, subtraction, multiplication, and division. Handle invalid input and division by zero.

class ChoiceRangeError(Exception):
    pass

def calculator():
    try:
        a = int(input("Enter 1st number: "))
        b = int(input("Enter 2nd number: "))
        print("Enter 1 for addition\nEnter 2 for subtraction\nEnter 3 for multiplication\nEnter 4 for division\n")

        choice = int(input("Enter your choice: "))
        if choice not in range(1,5):
            raise ChoiceRangeError("Invalid Input")
        elif choice in range(1,5):
            if choice == 1:
                return addition(a,b)
            elif choice == 2:
                return subtraction(a,b)
            elif choice == 3:
                return multiplication(a,b)
            elif choice == 4:
                return division(a,b)
    except ValueError:
        print("Error: Invalid Input")
    except ChoiceRangeError as e:
        print(f"Error: {e}")

def addition(num1,num2):
    print(num1+num2)

def subtraction(num1,num2):
    print(num1-num2) 

def multiplication(num1,num2):
    print(num1*num2)

def division(num1,num2):
    try:
        print(num1/num2)
    except ZeroDivisionError:
        print("Error: Can't do the division by zero")

calculator()
```

```python
# [ ] 20. Write a program that accepts a list of numbers separated by commas. Handle invalid numbers without crashing the entire program.

numbers = []

def num_list():
    while True:
        n = input("Enter the number to add in your list / (Enter'q' once done): ")
        if n.lower() == "q":
            break
        
        try:
            n = int(n)
            numbers.append(n)

        except ValueError:
            print("Error: Invalid Input")
    
    print(numbers)

num_list()
```
