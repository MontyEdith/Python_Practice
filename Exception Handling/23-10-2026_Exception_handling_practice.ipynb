```python
# [ ] 21. Write a function withdraw(balance, amount) that raises ValueError if the amount is zero or negative and raises an error if the withdrawal exceeds the balance.

class BalanceError(Exception):
    pass


def withdraw(balance, amount):
    if amount <= 0:
        raise ValueError("Amount can't be zero or negative")
    else:
        if amount > balance:
            raise BalanceError("You don't have enough balance")
        else:
            new_balance = balance - amount
            return new_balance
            

try:
    balance1 = withdraw(1000,-1200)
    print(f"Balance after withdrawal is {balance1}")
except ValueError as v:
    print(f"Error: {v}")
except BalanceError as b:
    print(f"Error: {b}")

```

```python
# [ ] 22. Create a custom exception called InvalidMarksError. Raise it if student marks are below 0 or above 100.

class InvalidMarksError(Exception):
    pass

def student(marks):
    if marks < 0 or marks > 100:
        raise InvalidMarksError("Marks should be in range 0-100")
    else:
        return f"The students marks is {marks}"

try:
    marks2 = student(10)
    print(marks2)
except InvalidMarksError as ime:
    print(f"Error: {ime}")

```

```python
# [ ] 23. Write a function that checks a password. Raise a custom WeakPasswordError if it has fewer than 8 characters.

class WeakPasswordError(Exception):
    pass

def password_check(password):
    if len(str(password)) < 8:
        raise WeakPasswordError("Weak password, it should be above 8 characters")
    return f"Strong password"

try:
    password = password_check(5656535243231)
    print(password)
except WeakPasswordError as e:
    print(f"Error: {e}")
```

```python
# [ ] 24. Create a student registration program that raises a custom DuplicateStudentError if a student ID already exists.

class DuplicateStudentError(Exception):
    pass

student_id = [101,102,103,104,105,106]
def student_registration(id):
    if id in student_id:
        raise DuplicateStudentError("This student ID already exists")

    return f"Registration done successfully"

try:
    id = student_registration(108)
    print(id)
except DuplicateStudentError as e:
    print(f"Error: {e}")
```

```python
# [ ] 25. Write a function that calculates the average of a list. Handle an empty list and non-numeric values appropriately.

class EmptyListError(Exception):
    pass

class NonNumericValuesError(Exception):
    pass

def avg_list(list1):
    if not list1:
        raise EmptyListError("List should not be empty")

    total = 0
    count = 0
    for i in list1:
        try:
            num = float(i)
        except:
            raise NonNumericValuesError("All values must be numeric")

        total += num
        count += 1

    return total/count

try:
    avg = avg_list([1,2,"jagf",4,5])
    print(avg)

except EmptyListError as e:
    print(f"Error: {e}")

except NonNumericValuesError as e:
    print(f"Error: {e}")

```
