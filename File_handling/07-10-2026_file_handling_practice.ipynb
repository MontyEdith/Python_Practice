```python
# Create a File
# Create a file named hello.txt and write "Hello, Python!" into it.

import os

file_name = "hello.txt"

filename_path = os.path.join(os.getcwd(), file_name)

f = open(filename_path, "w")

f.write("Hello, Python!")
f.close()
```

```python
# Read a File
# Read and display the complete contents of hello.txt.

import os

filepath = os.path.join(os.getcwd(), "hello.txt")

with open(filepath, "r") as f:
    content = f.read()
    print(content)
```

```python
# Read Line by Line
# Read about_me.txt line by line using a loop and print each line.

import os

filepath = os.path.join(os.getcwd(), "about_me.txt")

with open(filepath, "r") as f:
    for i in f:
        print(i)
```

```python
# Read First 10 Characters
# Create a file containing a paragraph and display only the first 10 characters.

import os

filepath = os.path.join(os.getcwd(), "para.txt")

with open(filepath, "w") as f:
    f.write("File handling in Python enables a program to create, read, update, \nand delete files stored permanently on a secondary storage device. \nThis process bridges the gap between volatile system memory (RAM) and non-volatile storage, ensuring that data persists\n even after the program terminates. Python executes these tasks using the built-in open(filename, mode) function, which\n accepts the file's path and a designated processing mode. The most common modes include 'r' for reading existing data, 'w' for writing new\n text (which entirely overwrites existing contents), and 'a' for appending new data to the end of the file without \ndeleting the original content")
```

```python
import os

filepath = os.path.join(os.getcwd(), "para.txt")

with open(filepath, "r") as f:
    print(f.read(10))
```

```python
# Append Data
# Open notes.txt in append mode and add a new note without deleting the existing content.

filepath = os.path.join(os.getcwd(), "notes.txt")

with open(filepath, "a") as f:
    f.write("\nTrue")

with open(filepath, "r") as f:
    print(f.read())
```

```python
# Write 5 Names
# Ask the user for 5 names and store them in names.txt, with each name on a separate line.

name_list = []
print("Please enter five names: ")

while len(name_list) < 5:
    name = input("Enter name: ")
    name_list.append(name)

filename = "names.txt"
filepath = os.path.join(os.getcwd(), filename)

with open(filepath, "w") as f:
    for i in name_list:
        f.write(f"{i}\n")

with open(filepath, "r") as f:
    print(f.read())
```

```python
import os

filename = "names.txt"
filepath = os.path.join(os.getcwd(), filename)

with open(filepath, "w") as f:
    for i in range(5):
        name = input("Enter name: ")
        f.write(f"{name}\n")

with open(filepath, "r") as f:
    print(f.read())
```

```python
# Display File Contents with Line Numbers
# Read names.txt and display:
# 1. Rahul
# 2. Priya
# 3. Aman

count = 1
with open(filepath, "r") as f:
    for i in f:
        print(f"{count}. {i}",end="")
        count += 1
```

```python
# Count Lines
# Read a text file and count how many lines it contains.

filepath = os.path.join(os.getcwd(),"para.txt")

count = 0
with open(filepath, "r") as f:
    for i in f:
        count += 1

print(f"Number of lines {"para.txt"} contatins is {count}")
```

```python
# Check Whether a File Exists
# Ask the user for a filename and check whether the file exists before attempting to read it.

import os

filename_input = input("Enter your file name: ")
filename = filename_input + ".txt"
filepath = os.path.join(os.getcwd(),filename)

if os.path.exists(filepath):
    with open(filepath, "r") as f:
        print(f.read())
else:
    print("File does not exist...")
```
