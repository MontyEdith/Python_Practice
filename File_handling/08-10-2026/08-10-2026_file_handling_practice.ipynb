```python
# Count Words
# Read a text file and count the total number of words.

import os

filename = "para.txt"
filepath = os.path.join(os.getcwd(), filename)

with open(filepath, "r") as f:
    content = f.read()

print(f"The total number of words are {len(content.split())}")
```

```python
# Count Vowels
# Read a file and count the number of vowels (a, e, i, o, u) present in it.

import os

filename = "para.txt"
filepath = os.path.join(os.getcwd(), filename)

with open(filepath, "r") as f:
    content = f.read().lower()

count = 0

for i in content:
    if i in "aeiou":
        count += 1

print(f"The number of vowels (a, e, i, o, u) present in {filename} are {count}")
```

```python
# Count Consonants
# Read a file and count the number of consonants.

import os

filename = "para.txt"
filepath = os.path.join(os.getcwd(), filename)

with open(filepath, "r") as f:
    content = f.read().lower()

count = 0

for i in content:
    if i not in "aeiou" and i.isalpha():
        count += 1

print(f"The number of consonants present in {filename} are {count}")
```

```python
# Count Digits
# Read a file containing text, numbers, and symbols and count how many digits it contains.

import os

filename = "para.txt"
filepath = os.path.join(os.getcwd(), filename)

with open(filepath, "r") as f:
    content = f.read().lower()

count = 0

for i in content:
    if i.isdigit():
        count += 1

print(f"The number of digits present in {filename} are {count}")
```

```python
# Find the Longest Line
# Read a file and find which line contains the maximum number of characters.

import os

filename = "para.txt"
filepath = os.path.join(os.getcwd(), filename)

with open(filepath, "r") as f:
    content = f.read()

largest = ""
for i in content.split("\n"):
    if len(i) > len(largest):
        largest = i

print("The line contains the maximum number of characters is")
print(largest)
print(len(largest))
```

```python
# Search for a Word
# Ask the user for a word and check whether that word exists in a file.

word = input("Enter the word you are searching for: ")

import os

filename = "para.txt"
filepath = os.path.join(os.getcwd(), filename)

with open(filepath, "r") as f:
    content = f.read().lower()

if word.lower() in content.split():
    print(f"{word} exist in {filename}")
else:
    print(f"{word} does not exist in {filename}")
```

```python
# Count Word Occurrences
# Ask the user for a word and count how many times it appears in a file.


import os
import string

word = input("Enter the word you are searching for: ")

filename = "para.txt"
filepath = os.path.join(os.getcwd(), filename)

with open(filepath, "r") as f:
    content = f.read().lower()

count = 0
words = content.split()

for i in words:
    i = i.strip(string.punctuation)
    
    if i == word.lower():
        count += 1

if count > 0:
    print(f"{word} appears {count} times in {filename}")
else:
    print(f"{word} does not exist in {filename}")
```

```python
# Convert Text to Uppercase
# Read a file and create another file containing the same content in uppercase.

import os

filename1 = "para.txt"
filepath1 = os.path.join(os.getcwd(), filename1)

with open(filepath1, "r") as f:
    content1 = f.read()

filename2 = "para_2.txt"
filepath2 = os.path.join(os.getcwd(),filename2)

with open(filepath2, "w") as f:
    f.write(content1.upper())

with open(filepath2, "r") as f:
    content2 = f.read()
    print(content2)
```

```python
# Remove Blank Lines
# Read a file and create a new file with all blank/empty lines removed.

import os

filename1 = "para.txt"
filepath1 = os.path.join(os.getcwd(), filename1)

filename2 = "para_3.txt"
filepath2 = os.path.join(os.getcwd(), filename2)

with open(filepath1, "r") as f1:
    with open(filepath2, "w") as f2:
        for line in f1:
            if line.strip() != "":
                f2.write(line)

print("Blank lines removed successfully!")
```

```python
# Copy One File to Another
# Copy all contents of source.txt into destination.txt.

source_path = os.path.join(os.getcwd(),"source.txt")
destination_path = os.path.join(os.getcwd(), "destination.txt")

with open(source_path, "r") as f1:
    with open(destination_path, "w") as f2:
        f2.write(f1.read())

```

```python
# Merge Two Files
# Read file1.txt and file2.txt and combine their contents into merged.txt.

file1_path = os.path.join(os.getcwd(),"source.txt")
file2_path = os.path.join(os.getcwd(), "para_3.txt")
merged_path = os.path.join(os.getcwd(), "merged.txt")

with open(file1_path, "r") as f1:
    with open(merged_path, "w") as f3:
        f3.write(f1.read())
        f3.write("\n")

with open(file2_path, "r") as f2:
    with open(merged_path, "a") as f3:
        f3.write(f2.read())

```

```python
# Separate Even and Odd Numbers
# Given a file containing numbers:
# 10
# 15
# 22
# 31
# 44
# Create:
# even.txt → even numbers
# odd.txt → odd numbers

import os

filename1 = "numbers.txt"
filepath1 = os.path.join(os.getcwd(), filename1)

filename2 = "even.txt"
filepath2 = os.path.join(os.getcwd(), filename2)

filename3 = "odd.txt"
filepath3 = os.path.join(os.getcwd(), filename3)


with open(filepath1, "r") as f1:
    content = f1.read().split()
    for i in content:
        if int(i)%2 == 0:
            with open(filepath2, "a") as f2:
                f2.write(f"{i} \n")
        else:
            with open(filepath3, "a") as f3:
                f3.write(f"{i} \n")

```

```python
# Find the Largest Number
# Read numbers from a file and find the largest number.

import os

filename1 = "numbers.txt"
filepath1 = os.path.join(os.getcwd(), filename1)

with open(filepath1, "r") as f1:
    content = f1.read().split()

largest = content[0]

for i in content:
    if int(i) > int(largest):
        largest = i
print(f"Largest numbe: {largest}")
```

```python
# Replace a Word
# Ask the user for:
# a word to search
# a replacement word
# Replace all occurrences in the file and save the updated content to a new file.

filename1 = "para.txt"
filepath1 = os.path.join(os.getcwd(), filename1)

filename2 = "new_para.txt"
filepath2 = os.path.join(os.getcwd(), filename2)

word_search = input("Enter the word you are searching for: ")
word_replace = input("Enter the word you want to replace with: ")

with open(filepath1, "r") as f1:
    content = f1.read()

content = content.replace(word_search, word_replace)

with open(filepath2, "w") as f2:
    f2.write(content)
```
