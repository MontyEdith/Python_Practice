```python
# Create a function is_positive() that checks whether a number is positive, negative, or zero.

def is_positive(num):

    if num > 0:
        print(f"{num} is positive")
    elif num < 0:
        print(f"{num} is negative")
    else:
        print(f"{num} is zero")

is_positive(657)
```

```python
# Create a function is_uppercase() that checks whether a character is uppercase.

def is_uppercase(char):

    if char == char.upper():
        print(f"{char} is uppercase")
    else:
        print(f"{char} is not uppercase")

is_uppercase("v")
```

```python
# Write a function is_digit() that checks whether a character is a digit.

def is_digit(char):

    if char.isdigit():
        print(f"{char} is digit")
    else:
        print(f"{char} is not digit")

is_digit("546")
```

```python
# Create a function check_age() that takes age and prints whether the person is eligible to vote.

def check_age(age):

    if age >= 18:
        print("Person is eligible to vote")
    else:
        print("Person is not eligible to vote")

check_age(31)
```

```python
# Write a function absolute(n) that returns the absolute value without using abs().

def absolute(n):
    if n < 0:
        return -n
    return n

absolute_value = absolute(-98)
print(f"Absolute: {absolute_value}")
```

```python
# Create a function average(a, b, c) that returns the average of three numbers.

def average(a,b,c):
    sum = a+b+c
    avg = sum/3
    return avg

num_average = average(2,8,20)
print(num_average)
```

```python
# Write a function calculate_percentage(total, obtained) that returns the percentage.

def calculate_percentage(total,obtained):
    fnal_per = (obtained/total)*100
    return fnal_per

percentage = calculate_percentage(400,285)
print(percentage)
```

```python
# Create a function calculate_simple_interest(p, r, t) that returns simple interest.

def calculate_simple_interest(p,r,t):
    sim_int = (p*r*t)/100
    return sim_int

simple_interest = calculate_simple_interest(5000,5,2)
print(simple_interest)
```

```python
# Write a function calculate_area(length, width) that returns the area of a rectangle.

def calculate_area(length, width):
    area = length*width
    return area

area_rectangle = calculate_area(10,20)
print(area_rectangle)
```

```python
# Create a function calculate_circle_area(radius) that returns the area of a circle.

import math

def calculate_circle_area(radius):
    area = math.pi * (radius**2)
    return f"{area:.2f}"

area_circle = calculate_circle_area(5)
print(area_circle)
```

```python
# Write a function calculate_triangle_area(base, height) that returns the area of a triangle.

def calculate_triangle_area(base, height):
    area = 0.5 * base * height
    return f"{area:.2f}"

area_triangle = calculate_triangle_area(5,10)
print(area_triangle)
```

```python
# Create a function calculate_temperature(celsius) that converts Celsius to Fahrenheit.

# Formula
# (0°C × 9/5) + 32 = 32°F


def calculate_temperature(celsius):
    fah = (celsius *(9/5)) + 32
    return f"{fah:.2f}"

temp_cel = calculate_temperature(37)
print(temp_cel)
```

```python
# Write a function calculate_temperature(fahrenheit) that converts Fahrenheit to Celsius.

# Formula
# (32°F − 32) × 5/9 = 0°C

def calculate_temperature(fahrenheit):
    cel =(fahrenheit - 32) * (5/9)
    return f"{cel:.2f}"

temp_fah = calculate_temperature(98.60)
print(temp_fah)
```

```python
# Create a function is_leap_year(year) that returns True or False.

def is_leap_year(year):
    if year%400 == 0 or (year%100 != 0 and year%4 == 0):
        return True
    return False

leap_year = is_leap_year(2024)
print(leap_year)
```

```python
# Write a function is_prime(n) that returns whether a number is prime.

def is_prime(n):
    prime = True

    if n <= 1:
        prime = False

    for i in range(2,n):
        if n%i == 0:
            prime = False
            break

    if prime:
        return f"{n} is a prime number"
    else:
        return f"{n} is not a prime number"

prime_num = is_prime(77)
print(prime_num)
```

```python
# Create a function factorial(n) that returns the factorial of a number.

def factorial(n):
    fact = 1

    for i in range(n,0,-1):
        fact *= i
    return fact

num_factorial = factorial(7)
print(num_factorial)
```
