```python
# Create a function is_palindrome(n) that checks whether a number is a palindrome.

def is_palindrome(n):
    num = n
    n = abs(n)
    reverse = 0

    while n>0:
        digit = n%10
        reverse = reverse*10 +digit
        n = n//10

    if reverse == num:
        return f"Palindrome : {num} : {reverse}"
    else:
        return f"Not a palindrome : {num} : {reverse}"
    

num_palindrome = is_palindrome(0)
print(num_palindrome)
```

```python
# Write a function is_armstrong(n) that checks whether a number is an Armstrong number.

def is_armstrong(n):
    n = abs(n)
    num = n
    power = 0
    total = 0

    while n > 0:
        power += 1
        n //=10

    n = num
    while n > 0:
        digit = n%10
        total = total + digit**power
        n //=10

    if total == num:
        return f"{num} is an Armstrong number : {total}"
    else:
        return f"{num} is not an Armstrong number : {total}"

arm_num = is_armstrong(153)
print(arm_num)

```

```python
# Create a function is_perfect(n) that checks whether a number is a perfect number.

def is_perfect(n):
    n = abs(n)
    total = 0

    for i in range(1,n):
        if n%i == 0:
            total += i

    if total == n:
        return f"{n} is a perfect number"
    else:
        return f"{n} is not a perfect number"

num_perf = is_perfect(-496)
print(num_perf)
```

```python
# Write a function is_strong(n) that checks whether a number is a Strong number.

def is_strong(n):
    n = abs(n)
    num = n
    total = 0

    while n > 0:
        digit = n%10
        fact = 1

        for i in range(1,digit+1):
            fact *= i

        total +=fact
        n //= 10

    if total == num:
        return f"{num} is a Strong number : {total}"
    else:
        return f"{num} is not a Strong number : {total}"

num_strong = is_strong(145)
print(num_strong)
```

```python
# Create a function is_harshad(n) that checks whether a number is a Harshad number.

def is_harshad(n):
    n = abs(n)
    num = n
    total = 0

    while n > 0:
        digit = n%10
        total += digit
        n //= 10

    if num%total == 0:
        return f"{num} is a Harshad number : {total}"
    else:
        return f"{num} is not a Harshad number : {total}"

num_harshad =  is_harshad(1729)
print(num_harshad)
```

```python
# Write a function gcd(a, b) that returns the GCD/HCF of two numbers.

def gcd(a,b):
    c = min(a,b)

    for i in range(c,0,-1):
        if a%i == 0 and b%i == 0:
            return i

    return 0
    
gcd_num = gcd(12,18)
print(gcd_num)
```

```python
# Create a function lcm(a, b) that returns the LCM of two numbers.

def lcm(a,b):
    a = abs(a)
    b = abs(b)

    if a == 0 or b == 0:
        return 0

    smaller = min(a,b)

    for i in range(smaller,0,-1):
        if a%i == 0 and b%i == 0:
            gcd_value = i
            break

    lcm_value = (a*b)//gcd_value
    return lcm_value

num_lcm = lcm(4,6)
print(num_lcm)
```

```python
# Write a function fibonacci(n) that returns the first n Fibonacci terms.

def fibonacci(n):
    n = abs(n)
    fib = []
    t = 0
    a = 0
    b = 1

    while t < n:
        fib.append(a)
        a , b = b, a+b
        t += 1

    return fib

fib_term = fibonacci(12)
print(fib_term)
```

```python
# Create a function prime_numbers(start, end) that returns all prime numbers between two numbers.

def prime_numbers(start, end):

    prime_num = []
    for i in range(start,end+1):
        if i < 2:
            continue 
        
        prime = True

        for j in range(2,i):
            if i%j == 0:
                prime = False

        if prime:
            prime_num.append(i)

    return prime_num

prime_num_bw_numbers = prime_numbers(1,100)
print(prime_num_bw_numbers)
```

```python
# Write a function is_anagram(text1, text2) that checks whether two strings are anagrams.

def is_anagram(text1, text2):
    string1 =""
    string2 = ""

    #Normalize both strings
    for i in text1.lower():
        if i != " ":
            string1 += i

    for j in text2.lower():
        if j != " ":
            string2 += j

    # Check the texts length
    string1_length = len(string1)
    string2_length = len(string2)

    if string1_length == string2_length:
        length_match = True
    else:
        length_match = False

    #Count character frequencies
    string1_freq1 = {}
    string2_freq2 = {}

    for i in string1:
        if i in string1_freq1:
            string1_freq1[i] += 1
        else:
            string1_freq1[i] = 1

    for j in string2:
        if j in string2_freq2:
            string2_freq2[j] += 1
        else:
            string2_freq2[j] = 1

    #Compare character frequencies

    if string1_freq1 == string2_freq2:
        char_freq = True
    else:
        char_freq = False

    if length_match and char_freq:
        return f"{text1} and {text2} are anagrams"
    else:
        return f"{text1} and {text2} are not anagrams"

anagrams = is_anagram("Tom Marvolo Riddle","I am Lord Voldemort")
print(anagrams)

    
```

```python
# Create a function rotate_list(numbers, k) that rotates a list by k positions.

def rotate_list(numbers, k):

    k = k%len(numbers)

    if k != 0:
        num_rotate = numbers[-k:] + numbers[:-k]
    else:
        num_rotate = numbers[:]

    return num_rotate

rot_k_list = rotate_list([1,2,3,4,5], 5)
print(rot_k_list)

    
```
