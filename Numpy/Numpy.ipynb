```python
# 93. Student Performance Analysis

#     Create a 2D NumPy array containing marks of 5 students across 5 subjects.

# Requirements:
# - Calculate total marks of each student.
# - Calculate the average marks of each student.
# - Find the highest-scoring student.
# - Find the lowest-scoring student.
# - Calculate the average marks of each subject.
# - Identify students who scored above 80 in all subjects.

marks = np.array([
    [85, 90, 78, 92, 88],
    [76, 82, 89, 80, 75],
    [92, 95, 91, 87, 94],
    [68, 74, 80, 72, 78],
    [88, 85, 90, 93, 86]
])
total_marks = np.sum(marks,axis = 1)
print(f"Total marks of each student: {total_marks}")
print("\n")
avg_marks = np.mean(marks,axis = 1)
print(f"Average marks of each student: {avg_marks}")
print("\n")
highest_scoring_student = np.argmax(total_marks) + 1
print(f"Highest-scoring studentt: {highest_scoring_student}")
print("\n")
lowest_scoring_student = np.argmin(total_marks) + 1
print(f"Lowest-scoring student: {lowest_scoring_student}")
print("\n")
avg_marks_sub = np.mean(marks,axis = 0)
print(f"Average marks of each student: {avg_marks_sub}")
print("\n")
condition = np.all(marks>80,axis=1)
students = np.where(condition)[0] + 1
print(f"Students who scored above 80 in all subjects: {students}")
        

```

```python
# 94. E-commerce Sales Analysis

# Create two matrices:
# - A matrix containing the quantity of products sold by 4 shops across 5 products.
# - A matrix containing the prices of those 5 products.
# Requirements:
# - Calculate total revenue generated by each shop.
# - Calculate revenue generated by each product.
# - Calculate the overall revenue.
# - Find the shop generating the highest revenue.
# - Find the most profitable product by revenue.
# Hint: Explore broadcasting and np.dot().

quantities = np.array([
    [10, 15, 20, 12, 8],   # Shop 1
    [12, 18, 16, 10, 14],  # Shop 2
    [8,  20, 25, 15, 10],  # Shop 3
    [15, 12, 18, 20, 16]   # Shop 4
])

prices = np.array([100, 250, 150, 300, 200])

total_revenue = quantities*prices
print(f"Total revenue generated by each shop: {np.sum(total_revenue,axis=1)}")
print(f"Total revenue generated by each product: {np.sum(total_revenue,axis=0)}")
print(f"Overall revenue generated: {np.sum(total_revenue)}")
shop_revenue = np.sum(total_revenue,axis=1)
print(f"Shop generating the highest revenue: {np.argmax(shop_revenue)+1}")
product_revenue = np.sum(total_revenue,axis=0)
print(f"Most profitable product by revenue: {np.argmax(product_revenue)+1}")
```
