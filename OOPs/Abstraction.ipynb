```python
# Beginner
#  Questions 1–4


# 1. Animal Sound System
# - Create an abstract class Animal.
# - Define an abstract method make_sound().
# - Create subclasses Dog and Cat.
# - Implement make_sound() in both subclasses.
# - Create objects and call their methods.

from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        print("Animal makes sound")

class Dog(Animal):

    def make_sound(self):
        print("Barks")


class Cat(Animal):

    def make_sound(self):
        print("Meows")

d1 = Dog()
d1.make_sound()

c1 = Cat()
c1.make_sound()
```

```python
# 2. Vehicle Rental System
# - Create an abstract class Vehicle.
# - Define an abstract method calculate_rent(days).
# - Create subclasses Car and Bike.
# - Calculate rental charges differently for each vehicle.

from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def calculate_rent(self,days):
        pass

class Car(Vehicle):

    def calculate_rent(self,days):
        value = days*100
        print(f"Car rent: {value}")

class Bike(Vehicle):

    def calculate_rent(self,days):
        value = days*50
        print(f"Bike rent: {value}")

vehicles = [Car(),Bike()]

for vehicle in vehicles:
    vehicle.calculate_rent(5)
```

```python
# Intermediate
#  Questions 5–8


# 5. Banking System
# - Create an abstract class BankAccount.
# - Define abstract methods deposit() and withdraw().
# - Create subclasses SavingsAccount and CurrentAccount.
# - Implement deposit and withdrawal operations.
# - Apply different withdrawal rules to both accounts.

from abc import ABC, abstractmethod

class BankAccount(ABC):

    @abstractmethod
    def deposit(self,amount):
        pass

    @abstractmethod
    def withdraw(self,amount):
        pass

class SavingsAccount(BankAccount):
    def __init__(self,balance):
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount
        print(f"Balance in Savings Account after depositing {amount} is {self.balance}")

    def withdraw(self,amount):
        self.balance -= amount
        print(f"Balance in Savings Account after withdrawing {amount} is {self.balance}\n")

class CurrentAccount(BankAccount):
    def __init__(self,balance):
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount
        print(f"Balance in Current Account after depositing {amount} is {self.balance}")

    def withdraw(self,amount):
        self.balance -= amount
        print(f"Balance in Current Account after withdrawing {amount} is {self.balance}")

accounts = [SavingsAccount(1000),CurrentAccount(2000)]

for account in accounts:
    account.deposit(500)
    account.withdraw(200)
```

```python
# 10. Mini Project – Food Delivery Management System

# Build a food delivery system using abstraction.

# Requirements:
# - Create an abstract class Delivery.
# - Define abstract methods:
#   - calculate_delivery_charge()
#   - assign_delivery_partner()
#   - track_order()
# - Create subclasses BikeDelivery, CarDelivery, and BicycleDelivery.
# - Implement different delivery charges based on distance.
# - Assign delivery partners.
# - Track delivery status.
# - Store multiple delivery objects in a list.
# - Process all deliveries using a loop.

from abc import ABC, abstractmethod

class Delivery(ABC):

    @abstractmethod
    def calculate_delivery_charge(self,charge):
        pass

    @abstractmethod
    def assign_delivery_partner(self,partner):
        pass

    @abstractmethod
    def track_order(self,partner):
        pass

class BikeDelivery(Delivery):

    def calculate_delivery_charge(self,distance):
        delivery_charge = 20 * distance
        print(f"Bike Delivery charge: {delivery_charge}")

    def assign_delivery_partner(self):
        print("Delivery partner for Bike delivery is PandaTraveller")

    def track_order(self):
        print("Order is in transit")

class CarDelivery(Delivery):

    def calculate_delivery_charge(self,distance):
        delivery_charge = 50 * distance
        print(f"Car Delivery charge: {delivery_charge}")

    def assign_delivery_partner(self):
        print("Delivery partner for Car delivery is TigressTraveller")

    def track_order(self):
        print("Order is in transit")

class BicycleDelivery(Delivery):

    def calculate_delivery_charge(self,distance):
        delivery_charge = 20 * distance
        print(f"Bicycle Delivery charge: {delivery_charge}")

    def assign_delivery_partner(self):
        print("Delivery partner for Bicycle delivery is MantisTraveller")

    def track_order(self):
        print("Order is in transit")

deliveries = [BikeDelivery(),CarDelivery(),BicycleDelivery()]

for delivery in deliveries:
    delivery.calculate_delivery_charge(500)
    delivery.assign_delivery_partner()
    delivery.track_order()
```
