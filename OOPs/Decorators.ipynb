```python
# 1. Create a Simple Decorator
# - Create a function my_decorator().
# - Define an inner function inside it.
# - Print "Before function execution" before calling the original function.
# - Print "After function execution" after calling the original function.
# - Apply the decorator to a function greet().

def my_decorator(func):

    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")

    return wrapper

@my_decorator
def greet():
    print("Hello Everyone")

greet()
```

```python
# 2. Function Execution Decorator
# - Create a decorator that prints "Function is starting".
# - Execute the original function.
# - Print "Function has finished" after execution.
# - Apply it to a function that displays your name.

def my_decorator(func):

    def wrapper():
        print("Function is starting")
        func()
        print("Function has finished")

    return wrapper

@my_decorator
def name():
    print("My name is Rishabh")

name()
```

```python
# 3. Decorator for Welcome Message
# - Create a decorator called welcome_decorator.
# - Print "Welcome to Python Programming!" before executing the original function.
# - Apply it to a function that displays "Learning Decorators".

def welcome_decorator(func):

    def wrapper():
        print("Welcome to Python Programming!")
        func()
        print("Learning Decorators")

    return wrapper

@welcome_decorator
def welcome_message():
    print("Welcome to real word! It sucks, you gonna love it")

welcome_message()
```

```python
# 4. Decorator for Multiplication
# - Create a decorator that prints "Performing multiplication...".
# - Create a function multiply(a, b) that returns their product.
# - Apply the decorator to the function.

def product_decorator(func):

    def wrapper(a,b):
        print("Performing multiplication...")
        func(a,b)
        print("Multiplication done")

    return wrapper

@product_decorator
def multiply(a,b):
    prod = a*b
    print(f"Multiplication: {prod}")

multiply(2,3)
```

```python
# 5. Decorator for Uppercase Conversion
# - Create a decorator that converts the returned string of a function into uppercase.
# - Create a function greet() that returns "hello, welcome to python".
# - Apply the decorator.

def uppercase_conversion(func):

    def wrapper():
        print("Decorator for Uppercase Conversion!")
        x = func()
        return x.upper()

    return wrapper

@uppercase_conversion
def greet():
    return "hello, welcome to python"

result = greet()
print(result)
```

```python
# 6. Decorator for Lowercase Conversion
# - Create a decorator that converts the returned string into lowercase.
# - Apply it to a function that returns "PYTHON DECORATORS".

def lowercase_conversion(func):

    def wrapper():
        print("Decorator for Lowercase Conversion!")
        x = func()
        return x.lower()

    return wrapper

@lowercase_conversion
def greet():
    return "PYTHON DECORATORS"

result = greet()
print(result)
```

```python
# 7. Decorator for Function Name
# - Create a decorator that prints the name of the function being executed.
# - Apply it to three different functions.
# - Display the function names whenever they are called.

def function_name(func):

    def wrapper():
        print("Decorator for Function Name!")
        print(f"Executing function {func.__name__}")
        func()

    return wrapper

@function_name
def greet():
    print("Hello Everyone\n")

@function_name
def welcome_message():
    print("Welcome to real word! It sucks, you gonna love it\n")

@function_name
def name():
    print("My name is Rishabh\n")

greet()
welcome_message()
name()
```

```python
# 8. Decorator for Execution Count
# - Create a decorator that counts how many times a function has been executed.
# - Call the decorated function five times.
# - Display the total number of executions.

count = 0
def execution_count(func):

    def wrapper():
        global count
        count += 1
        func()
        print(f"Function got executed {count} times\n")

    return wrapper

@execution_count
def name():
    print("My name is Rishabh")

name()
name()
name()
name()
name()
```

```python
# 9. Decorator for Fixed Message
# - Create a decorator that prints "This function is protected by a decorator".
# - Apply it to a function that displays your personal details.
# - Execute the decorated function.

def my_decorator(func):

    def wrapper():
        print("This function is protected by a decorator")
        func()
        print("Decorator ended!")

    return wrapper

@my_decorator
def name():
    print("My name is Rishabh")
    print("I am an AI Engineer")
    print("I lives in Bangalore")

name()

```

```python
# 10. Decorator for Return Value
# - Create a decorator that adds 100 to the return value of a function.
# - Create a function get_salary() that returns 5000.
# - Apply the decorator and display the final salary.

def return_value(func):

    def wrapper():
        print(f"Decorator for Return Value")
        x = func()
        return x + 100

    return wrapper

@return_value
def get_salary():
    return 5000

final_salary = get_salary()
print(f"The final salary is {final_salary}")
```

```python
# 11. Decorator with Arguments
# - Create a decorator that works with functions accepting arguments.
# - Create a function add(a, b).
# - Print the arguments before executing the function.
# - Display the result after execution.


def my_decorator(func):

    def wrapper(a,b):
        print(f"First value: {a}")
        print(f"Second value: {b}")
        result = func(a,b)
        print(f"Addition: {result}")
        return result

    return wrapper

@my_decorator
def add(a,b):
    value = a + b
    return value

result = add(2,3)
```

```python
# *12. Decorator Using args
# - Create a decorator that accepts any number of positional arguments.
# - Create a function calculate_sum(*args).
# - Return the sum of all arguments.
# - Display the result.

def decorator_args(func):

    def wrapper(*args):
        for index,value in enumerate(args,start = 1):
            print(f"The {index} value is {value}")
        result = func(*args)
        # print(f"Addition: {result}")
        return result

    return wrapper

@decorator_args
def calculate_sum(*args):
    total = 0
    for i in args:
        total += i
    return total

result = calculate_sum(9,8,7,6,5,12,23,24,35,46)
print(f"Returned addition: {result}")
```

```python
# **13. Decorator Using kwargs
# - Create a decorator that accepts keyword arguments.
# - Create a function student_details(**kwargs).
# - Display all student details.
# - Print the keyword arguments before function execution.

def decorator_args(func):

    def wrapper(**kwargs):
        print("Student's details: ")
        for key,value in kwargs.items():
            print(f"The student's {key} is {value}")
            
        print("\nExeciting main function")
        func(**kwargs)

    return wrapper

@decorator_args
def student_details(**kwargs):
    for k,v in kwargs.items():
        print(f"{k} : {v}")

student_details(Name = "Rishabh", Age = 34, Location = "Pune")
```

```python
# 15. Login Authentication Decorator
# - Create a decorator called login_required.
# - Allow a function to execute only when is_logged_in = True.
# - Otherwise, display "Please login first!".
# - Apply it to a function dashboard().

def login_required(func):
    sys_pin = 1234

    def wrapper():
        count = 3

        while count>0:
            pin = int(input("Enter the pin: "))
            if pin == sys_pin:
                func()
                break
            else:
                print("Please login first!")
                count -= 1
                print(f"{count} attempts left")

    return wrapper

@login_required
def dashboard():
    print("Login Successful")

dashboard()
```

```python
# 16. Positive Number Validation
# - Create a decorator that checks whether the arguments passed to a function are positive.
# - If all arguments are positive, execute the function.
# - Otherwise, display an appropriate error message.
# - Apply it to a multiplication function.

def pos_num_val(func):

    def wrapper(*args):
        for i in args:
            if i < 0:
                print("Negative numbers are not allowed")
                return
            
        result = func(*args)
        return result

    return wrapper

@pos_num_val
def mul(*args):
    prod = 1
    for i in args:
        prod *= i
    return prod

result = mul(-2,3,4)
print(result)
```

```python
# 17. Even Number Validation
# - Create a decorator that checks whether the argument passed to a function is even.
# - Execute the function only if the number is even.
# - Otherwise, display "Please provide an even number".

def even_num_val(func):

    def wrapper(*args):
        for i in args:
            if i%2 != 0:
                print("Please provide an even number")
                return
            
        result = func(*args)
        return result

    return wrapper

@even_num_val
def even_num(*args):
    prod = []
    for i in args:
        prod.append(i)
    return prod

print(even_num(2, 4, 6, 8))
print(even_num(2, 4, 7, 8))
```

```python
# 18. Multiple Decorators
# - Create two decorators:
#   - uppercase_decorator
#   - exclamation_decorator
# - Create a function that returns "hello python".
# - Apply both decorators to the function.
# - Observe the order of execution.

def uppercase_decorator(func):

    def wrapper():
        value = func()
        value = value.upper()
        return value

    return wrapper

def exclamation_decorator(func):

    def wrapper():
        value = func()
        value += "!"
        return value

    return wrapper

@uppercase_decorator
@exclamation_decorator
def greet():
    return "hello python"

print(greet())

```

```python
# 19. Decorator with Parameters
# - Create a decorator that accepts a message as an argument.
# - Display that message before executing the original function.
# - Apply it to different functions with different messages.

def my_decorator(msg):
    def decorator(func):

        def wrapper(*args):
            print(msg)
            result = func(*args)
            return result

        return wrapper

    return decorator

@my_decorator("Multiplication:")
def mul(a,b):
    return a*b

@my_decorator("\nAddition:")
def add(a,b):
    return a+b

print(mul(2,3))

print(add(2,3))
```

```python
# 20. Repeat Function Decorator
# - Create a decorator that executes a function multiple times.
# - Accept the repetition count as a decorator argument.
# - Apply it to a function that prints "Learning Python".

def my_decorator(count):
    def decorator(func):

        def wrapper():
            print("Repeat Function Decorator")
            nonlocal count
            while count>0:
                func()
                count -= 1

        return wrapper

    return decorator

@my_decorator(5)
def repeat_function():
    print("Learning Python")

print(repeat_function())
```

```python
# 21. Exception Handling Decorator
# - Create a decorator that handles exceptions raised by a function.
# - Use try-except.
# - Display an appropriate error message if an exception occurs.
# - Apply it to a division function.

def except_handling_decorator(func):

    def wrapper(a,b):
        try:
            value = func(a,b)
            return value
        except Exception as e:
            print(f"Error: {e}")

    return wrapper

@except_handling_decorator
def div(a,b):
    divsion = a/b
    return divsion


print(div(20,0))
```

```python
# 22. Retry Decorator
# - Create a decorator that retries a function if it fails.
# - Allow a maximum of three attempts.
# - Display the attempt number.
# - Display a message if all attempts fail.

def login_required(func):
    sys_pin = 1234

    def wrapper():
        print("Welcome to Dashboard")

        count = 3
        while count>0:
            pin = int(input("Enter the pin: "))
            if pin == sys_pin:
                value = func()
                return value
            else:
                count -= 1
                print(f"{count} attempts left")
        print("Login Unsuccessful")

    return wrapper

@login_required
def dashboard():
    print("Login Successful")

dashboard()
```
