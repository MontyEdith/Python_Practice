```python
# 4. Getter and Setter: Create a BankAccount class with a private attribute __balance. Implement methods to deposit money, withdraw money, and check the balance.

class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
            print(f"Balance after deposit: {self.__balance}")
        else:
            print("Deposit amount should be greater than 0")

    def withdraw(self,amount):
            if amount > 0:
                if amount > self.__balance:
                    print("You don't have anough balance")
                else:
                    self.__balance -= amount
                    print(f"Balance after withdrawal: {self.__balance}")
            else:
                print("Withdrawal amount should be greater than 0")

    def check_balance(self):
        print(f"Your current balance is {self.__balance}")

ba_obj = BankAccount(1000)
ba_obj.check_balance()
ba_obj.deposit(500)
ba_obj.withdraw(1200)
ba_obj.check_balance()

```

```python
# 5. Protected Attributes: Create a Vehicle class with a protected attribute _brand. Create a Car subclass and access the protected attribute.

class Vehicle:
    def __init__(self,brand):
        self._brand = brand

class Car(Vehicle):
    def __init__(self, brand):
        super().__init__(brand)

    def get_brand(self):
        print(f"Car brand: {self._brand}")

c1 = Car("Hyundai")
c1.get_brand()
```

```python
# 6. Private Methods: Create a Calculator class with a private method __multiply(). Create a public method calculate() that internally calls the private method.

class Calculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b

    def __multiply(self):
        return self.a*self.b

    def calculate(self):
        print(f"Multiplication: {self.__multiply()}")

c1 = Calculator(3,8)
c1.calculate()
```

```python
# 7. Data Hiding: Create a User class with private attributes __username and __password. Allow users to retrieve their username but never expose their password directly.

class User:
    def __init__(self,username,password):
        self.__username = username
        self.__password = password

    def get_username(self):
        print(f"Username: {self.__username}")

    def get_password(self):
        print(f"Password: {len(self.__password)* "*"}")

u_obj = User("Edith","NoLook")
u_obj.get_username() 
u_obj.get_password()
```

```python
# 14. Password Management: Create a UserAccount class with a private password. Implement:
# - Set password.
# - Change password.
# - Verify password.
# - Ensure the new password contains at least 8 characters.

class UserAccount:
    def __init__(self,password):
        if len(password) >= 8:
            self.__password = password
        else:
            print("Please enter the password which contains at least 8 characters")

    def set_password(self,new_password):
        if len(new_password) >= 8:
            self.__password = new_password
            print(f"Changed Password: {self.__password}")
        else:
            print("Please enter the new password which contains at least 8 characters")

    def verify_password(self,password_check):
        if len(password_check) >= 8:
            if self.__password == password_check:
                print(f"Password verified, It's a match")
            else:
                print(f"Password verified, It's not a match")
        else:
            print("Please enter the password check which contains at least 8 characters")

ua1 = UserAccount("jgaddfgjkfhf")
ua1.verify_password("jgaddfgjkfhf")
ua1.verify_password("kdfdbffv")
ua1.set_password("gbavjhb7df")
```

```python
# 21. Use of @property: Create a Circle class with a private attribute __radius. Use @property to retrieve the radius and calculate the area. Use a setter to validate the radius.

class Circle:
    def __init__(self, radius):
        self.__radius = radius

    @property
    def radius(self):
        return self.__radius

    @radius.setter
    def radius(self, new_radius):
        if new_radius > 0:
            self.__radius = new_radius
        else:
            print("Radius must be greater than 0")

    @property
    def area(self):
        return 3.14159 * self.__radius ** 2


c1 = Circle(5)

print("Radius:", c1.radius)
print("Area:", c1.area)

c1.radius = 10

print("Updated Radius:", c1.radius)
print("Updated Area:", c1.area)

    
```

```python
# 23. Employee Performance: Create an Employee class with private attributes:
# - __name
# - __salary
# - __performance_rating
#     Allow salary increments only if the performance rating is greater than or equal to 4.

class Employee:
    def __init__(self,name,salary,performance_rating):
        self.__name = name
        self.__salary = salary
        self.__performance_rating = performance_rating

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self,hike):
        if self.__performance_rating >= 4:
            self.__salary += hike
            return self.__salary
        else:
            return f"{self.__name}, No salary increment for you"

emp1 = Employee("Monty",1000,8)
print(f"Employee salary: {emp1.salary}")
emp1.salary = 200
print(f"Employee salary after hike: {emp1.salary}")
print("\n")
emp2 = Employee("Rishabh",1000,2)
print(f"Employee salary: {emp2.salary}")
emp1.salary = 200
print(f"Employee salary after hike: {emp2.salary}")
```

```python
# 24. Digital Wallet: Create a Wallet class with private attributes __owner and __balance. Implement:
# - Add money.
# - Spend money.
# - Transfer money to another wallet.
# - Display wallet balance.


class Wallet:
    def __init__(self,owner,balance):
        self.__owner = owner
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    def add_money(self,amount):
        self.__balance += amount
        return self.__balance

    def spend_money(self,amount):
        self.__balance -= amount
        return self.__balance

    def transfer_money(self,receiver,amount):
        self.__balance -= amount
        receiver.__balance += amount

w1 = Wallet("Monty",1000)
w2 = Wallet("Rishabh",1000)

print(w1.balance)
print(w2.balance)

print(w1.add_money(100))

print(w1.spend_money(300))


w1.transfer_money(w2,500)

print(w1.balance)
print(w2.balance)
```

```python
# 25. Car Speed Control: Create a Car class with a private attribute __speed. Implement methods to accelerate and brake. Ensure speed never goes below 0 or exceeds 200 km/h.

class Car:
    def __init__(self,speed):
        if speed >= 0:
            self.__speed = speed
        else:
            print(f"Speed should not be a negative value")

    @property
    def speed(self):
        return f"Speed: {self.__speed}kmph"

    def accelerate_brake(self):
        if self.__speed >= 0:
            if self.__speed  < 200:
                return f"Speed is below 200kmph, keep accelerating"
            else:
                return f"Speed should not exceeds 200kmph, put on some brake"
        else:
            return f"Speed should not be a negative value"

c1 = Car(20)
print(c1.speed)
print(c1.accelerate_brake())
```

```python
# 30. Mini Project – Employee Payroll Management System

#     Build a complete employee payroll system using encapsulation.

# Requirements:
# - Create an Employee class.
# - Private attributes: __employee_id, __name, __basic_salary, __bonus, __deductions.
# - Create getter and setter methods.
# - Validate salary, bonus, and deductions.
# - Calculate gross salary.
# - Calculate net salary.
# - Prevent unauthorized modification of employee information.
# - Display complete employee details and salary slip.

class Employee:
    def __init__(self,employee_id,name,basic_salary,bonus,deductions):
        self.__employee_id = employee_id
        self.__name = name
        self.__basic_salary = basic_salary
        self.__bonus = bonus
        self.__deductions = deductions

    def get_id_name(self):
        print(f"Employee ID: {self.__employee_id}")
        print(f"Name: {self.__name}")

    def validate_salary(self,amount):
        if self.__basic_salary == amount:
            print(f"Salary validated, it's a match")
        else:
            print(f"Salary validated, it's not a match")

    def validate_bonus(self,amount):
        if self.__bonus == amount:
            print(f"Bonus validated, it's a match")
        else:
            print(f"Bonus validated, it's not a match")

    def validate_deductions(self,amount):
        if self.__deductions == amount:
            print(f"Deductions validated, it's a match")
        else:
            print(f"Deductions validated, it's not a match")

    def calculate_gross_salary(self):
        self.gross_salary = self.__basic_salary + self.__bonus
        print(f"Gross Salary: {self.gross_salary }")

    def calculate_net_salary(self):
            self.net_salary = self.gross_salary - self.__deductions
            print(f"Net Salary: {self.net_salary }")

    def employee_details(self,password):
            if password == 1234:
                print("-------" * 8)
                print(f"Employee ID: {self.__employee_id}")
                print(f"Employee name: {self.__name}")
                print("-------" * 8)
                print("Employee Salary slip")
                print(f"Employee basic salary: {self.__basic_salary}")
                print(f"Employee bonus: {self.__bonus}")
                print(f"Employee deductions: {self.__deductions}")
                print(f"Employee gross salary: {self.gross_salary}")
                print(f"Employee net salary: {self.net_salary}")
                print("-------" * 8)
            else:
                print("You are not authorised to see or modify the employee")

emp1 = Employee(107,"Monty",1000,200,100)
emp1.get_id_name()
emp1.validate_salary(100)
emp1.validate_salary(1000)
emp1.validate_bonus(200)
emp1.validate_bonus(1000)
emp1.validate_deductions(1000)
emp1.validate_deductions(100)
emp1.calculate_gross_salary()
emp1.calculate_net_salary()
emp1.employee_details(12345)
emp1.employee_details(1234)

```
