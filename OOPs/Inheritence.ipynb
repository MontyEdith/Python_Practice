```python
# 15. Student Results: Create a Student class with five subject marks. Create a Result subclass that calculates total marks, percentage, and pass/fail status.

class Student:
    def __init__(self,maths,physics,chemistry,english,hindi):
        self.maths = maths
        self.physics = physics
        self.chemistry = chemistry
        self.english = english
        self.hindi = hindi

class Result(Student):
    def __init__(self,maths,physics,chemistry,english,hindi):
        super().__init__(maths,physics,chemistry,english,hindi)
        self.total = self.maths + self.physics + self.chemistry + self.english + self.hindi
        self.percent = (self.total/500) * 100

    def total_marks(self):
        print(f"Total marks out of 500 is {self.total}")

    def percentage(self):
        print(f"Student's final percentage is {self.percent:.2f}")

    def status(self):
        marks = [self.maths, self.physics, self.chemistry, self.english, self.hindi]
        if all(mark >=40 for mark in marks):
            print(f"Student passed the exam")  
        else:
            print(f"Student failed the exam")

stu1 = Result(60,25,16,28,16)
stu1.total_marks()
stu1.percentage()
stu1.status()
```

```python
# 17. Online Shopping: Create a Product class with name and price. Create a DiscountedProduct subclass that accepts a discount percentage and calculates the final price.

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class DiscountedProduct(Product):
    def __init__(self,name, price,discount_percent):
        super().__init__(name, price)
        self.discount_percent = discount_percent

    def final_price(self):
        discount_amount = self.price*(self.discount_percent/100)
        final = self.price - discount_amount
        print(f"The final price of the product {self.name} is {final}")

p1 = DiscountedProduct("Aobing", 100, 20)
p1.final_price()
```

```python
# 36. Hybrid Inheritance: Design a class hierarchy combining multilevel and multiple inheritance using Person, Student, Athlete, and ScholarAthlete.

class Person:
    def __init__(self,name):
        self.name = name

class Student(Person):
    def __init__(self,name,school):
        super().__init__(name)
        self.school = school

    def display_student_details(self):
        print(f"Student name: {self.name} and student studies in {self.school}")

class Athelete(Student):
    def __init__(self,name,school,level):
        super().__init__(name,school)
        self.level = level

    def display_athelete_details(self):
        print(f"Student name: {self.name} and student studies in {self.school}") 
        print(f"Student is a {self.level} level athlete")

class ScholarAthlete(Student,Person):
    def __init__(self,name,school,level,percentage):
        Person.__init__(self,name)
        Student.__init__(self,name,school)
        self.level = level
        self.percentage = percentage

    def display_scholarathelete_details(self): 
        print(f"{self.name} is a scholar athlete as he scored {self.percentage} in {self.school} and won the medal at {self.level} level")

sa1 = ScholarAthlete("Monty","KV","national",75)
sa1.display_scholarathelete_details()
print("\n")
sa2 = Athelete("Rishabh","DAV","zonal")
sa2.display_athelete_details()
```

```python
# 41. Method Resolution Order (MRO): Create classes A, B(A), C(A), and D(B, C). Define a method named show() in every class. Create a D object and 
# investigate which method is called and why.

class A:
    def show(self):
        print("Class A")

class B(A):
    def show(self):
        print("Class B")

class C(A):
    def show(self):
        print("Class C")

class D(B,C):
    def show_d(self):
        print("Class D")

d = D()
d.show()

print(D.mro())
```

```python
# 43. Using isinstance(): Create a Person class and a Student subclass. Test whether objects of both classes are instances of Person and Student.

class Person:
    def display(self):
        print("Person class")

class Student(Person):
    def display(self):
        print("Student class")

per_obj = Person()
stu_obj = Student()

print(isinstance(per_obj, Person))
print(isinstance(per_obj, Student))
print(isinstance(stu_obj, Person))
print(isinstance(stu_obj, Student))
```

```python
# 44. Using issubclass(): Create Animal, Mammal, Bird, and Dog classes. Use issubclass() to verify their inheritance relationships.

class Animal:
    def display(self):
        print("Animal class")

class Mammal(Animal):
    def display(self):
        print("Mammal class")

class Bird(Animal):
    def display(self):
        print("Bird class")

class Dog(Mammal):
    def display(self):
        print("Dog class")


print(issubclass(Mammal,Animal))
print(issubclass(Bird,Animal))
print(issubclass(Dog,Animal))
print(issubclass(Animal,Mammal))
print(issubclass(Dog,Mammal))
print(issubclass(Mammal,Dog))
```

```python
# 50. Mini Project — Employee Management System: Build an employee management system using inheritance. Include:
# - Employee as the base class.
# - Developer, Manager, and Intern as subclasses.
# - Shared attributes: employee ID, name, and base salary.
# - Different salary calculations for each subclass.
# - Method overriding for displaying employee details.
# - super() for constructor reuse.
# - A list of employees and a loop that displays each employee's details and salary.


class Employee:
    def __init__(self,employee_ID,name,base_salary):
        self.employee_ID = employee_ID
        self.name = name
        self.base_salary = base_salary

    def employee_details(self):
        print(f"Employee name: {self.name}")
        print(f"Employee ID: {self.employee_ID}")
        print(f"Employee base salary: {self.base_salary}")

class Developer(Employee):
    def __init__(self,employee_ID,name,base_salary,salary_differ):
        super().__init__(employee_ID,name,base_salary)
        self.salary_differ = salary_differ

    def employee_details(self):
        print(f"Employee name: {self.name}")
        print(f"Employee ID: {self.employee_ID}")
        print(f"Employee base salary: {self.base_salary + self.salary_differ}")

class Manager(Employee):
    def __init__(self,employee_ID,name,base_salary,salary_differ):
        super().__init__(employee_ID,name,base_salary)
        self.salary_differ = salary_differ

    def employee_details(self):
        print(f"Employee name: {self.name}")
        print(f"Employee ID: {self.employee_ID}")
        print(f"Employee base salary: {self.base_salary + self.salary_differ}")

class Intern(Employee):
    def __init__(self,employee_ID,name,base_salary,salary_differ):
        super().__init__(employee_ID,name,base_salary)
        self.salary_differ = salary_differ

    def employee_details(self):
        print(f"Employee name: {self.name}")
        print(f"Employee ID: {self.employee_ID}")
        print(f"Employee base salary: {self.base_salary + self.salary_differ}")

employees = [Developer(101,"Rishabh",10000,1000),Manager(207,"Monty",10000,2000),Intern(309,"Ronny",10000,500)]

for employee in employees:
    employee.employee_details()
    print("\n")
```
