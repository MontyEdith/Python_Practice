```python
# 1. Animal Sounds
#    Create an Animal class with a make_sound() method. Create Dog, Cat, and Cow subclasses that override this method with their respective sounds.

class Animal:
    def make_sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def make_sound(self):
        print("Barks")

class Cat(Animal):
    def make_sound(self):
        print("Meows")

class Cow(Animal):
    def make_sound(self):
        print("Moos")

animals = [Dog(),Cat(),Cow()]

for animal in animals:
    animal.make_sound()
```

```python
# 2. Vehicle Movement
#    Create a Vehicle class with a move() method. Create Car, Bike, and Airplane subclasses with different implementations of move().

class Vehicle:
    def move(self):
        print("Vehicle moves")

class Car(Vehicle):
    def move(self):
        print("Car moves")

class Bike(Vehicle):
    def move(self):
        print("Bike runs")

class Airplane(Vehicle):
    def move(self):
        print("Airplane flies")

vehicles = [Car(),Bike(),Airplane()]

for vehicle in vehicles:
    vehicle.move() 
```

```python
# 3. Shape Area
#    Create a Shape class with an area() method. Override it in Rectangle, Circle, and Triangle classes to calculate their respective areas.

class Shape:
    def area(self):
        print("Checks area of the shape")

class Rectangle(Shape):
    def area(self):
        print("Area of rectangle = length * width")

class Circle(Shape):
    def area(self):
        print("Area of circle = pi * (radius **2)")

class Triangle(Shape):
    def area(self):
        print("Area of triangle = 0.5 * base * height")

shapes = [Rectangle(),Circle(),Triangle()]

for shape in shapes:
    shape.area() 
```

```python
# 22. Shape Area Calculator
#     Create Circle, Rectangle, and Square classes. Write a function calculate_area(shape) that calculates the area without checking the object's class.

class Circle:
    def area(self):
        print("Circle area = pi * (radius ** 2)")

class Rectangle:
    def area(self):
        print("Rectangle area = length * width")

class Square:
    def area(self):
        print("Square area = side ** 2")

def calculate_area(shape):
    for i in shape:
        i.area()

shapes = [Circle(),Rectangle(),Square()]
calculate_area(shapes)
```

```python
# 24. Salary Management
#     Create Manager, Developer, and Intern classes with a calculate_salary() method. Store their objects in a list and display their salaries using a loop.

class Manager:
    def __init__(self,base_salary,bonus,deductions):
        self.base_salary = base_salary
        self.bonus = bonus
        self.deductions = deductions

    def calculate_salary(self):
        salary = self.base_salary + self.bonus - self.deductions
        print(f"Manager's salary is {salary}")

class Developer:
    def __init__(self,base_salary,bonus,deductions):
        self.base_salary = base_salary
        self.bonus = bonus
        self.deductions = deductions

    def calculate_salary(self):
        salary = self.base_salary + self.bonus - self.deductions
        print(f"Developer's salary is {salary}")

class Intern:
    def __init__(self,base_salary,bonus,deductions):
        self.base_salary = base_salary
        self.bonus = bonus
        self.deductions = deductions

    def calculate_salary(self):
        salary = self.base_salary + self.bonus - self.deductions
        print(f"Intern's salary is {salary}")


employees = [Manager(5000,1000,400),Developer(3000,800,200),Intern(1000,500,100)]
for employee in employees:
    employee.calculate_salary()
```

```python
# 25. Payment Processing
#     Create different payment classes with a process_payment(amount) method. Write a common function that accepts any payment object and processes the payment.

class Payment:
    def __init__(self,amount):
        self.amount = amount

    def process_payment(self):
        print("Payment process done")

class GPay(Payment):
    def __init__(self,amount):
        super().__init__(amount)

    def process_payment(self):
        print(f"Payment of {self.amount} done via GPay")

class CreditCard(Payment):
    def __init__(self,amount):
        super().__init__(amount)

    def process_payment(self):
        print(f"Payment of {self.amount} done via Credit Card")

class DebitCard(Payment):
    def __init__(self,amount):
        super().__init__(amount)

    def process_payment(self):
        print(f"Payment of {self.amount} done via Debit Card")


def payment(payments):
    for i in payments:
        i.process_payment()

pay_obj1 = [GPay(20)]
pay_obj2 = [GPay(10),CreditCard(20)]
pay_obj3 = [GPay(30),CreditCard(70),DebitCard(50)]

payment(pay_obj1)
payment(pay_obj2)
payment(pay_obj3)
```

```python
# 26. Duck Typing – Media Player
#     Create AudioPlayer, VideoPlayer, and PodcastPlayer classes. Each class should have a play() method. Write a function that accepts any object and calls play().

class AudioPlayer:
    def play(self):
        print("AudioPlayer is playing")

class VideoPlayer:
    def play(self):
        print("VideoPlayer is playing")

class PodcastPlayer:
    def play(self):
        print("PodcastPlayer is playing")

def play_status(status):
    for i in status:
        i.play()

play_obj = [AudioPlayer(),VideoPlayer(),PodcastPlayer()]
play_status(play_obj)
```

```python
# 31. Add Two Objects
#     Create a Vector class with x and y coordinates. Overload the + operator using __add__() to add two vector objects.

class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __add__(self,other):
        x = self.x + other.x
        y = self.y + other.y
        return Vector(x,y)

    def __str__(self):
        return f"Result: ({self.x},{self.y})"

v1 = Vector(1,2)
v2 = Vector(3,4)
print(v1+v2)
```

```python
# 32. Compare Two Objects
#     Create a Student class with name and marks. Overload the > operator using __gt__() to compare two students based on marks.

class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

    def __gt__(self,other):
        return self.marks > other.marks

s1 = Student("Rishabh",23)
s2 = Student("Monty",45)
print(s1<s2)
        
```

```python
# 34. Object Equality
#     Create an Employee class with employee ID and name. Overload == using __eq__() to compare employees based on employee ID.

class Employee:
    def __init__(self,id,name):
        self.id = id
        self.name = name

    def __eq__(self,other):
        return self.id == other.id

e1 = Employee(12,"Monty")
e2 = Employee(12,"Rishabh")

print(e1==e2)
```

```python
# 35. Multiply Objects
#     Create a Product class with price and quantity. Overload the * operator using __mul__() to calculate the total product value.

class Product:
    def __init__(self,price,quantity):
        self.price = price
        self.quantity = quantity

    def __mul__(self,other):
        return self.price * other

p1 = Product(20,4)
print(p1 * p1.quantity)
```

```python
# 37. Custom Indexing
#     Create a Team class containing a list of players. Implement __getitem__() to access players using indexing.

class Team:
    def __init__(self,players):
        self.players = players

    def __getitem__(self,index):
        return self.players[index]

t1 = Team(["Rishabh","Monty","Saikat","King","Ajay","Badshah"])
print(t1[3])
```

```python
# 38. Method Overloading Using Default Arguments
#     Create a Calculator class with an add() method that can accept two or three numbers. Implement this using default arguments.

class Calculator:
    def add(self,a,b,c=0):
        return a + b + c

cal1 = Calculator()
print(cal1.add(2,3))
print(cal1.add(2,3,4))
print(cal1.add(4,7))
```

```python
# 39. Method Overloading Using Variable Arguments
#     Create a Calculator class with a multiply() method that accepts any number of arguments using *args.

class Calculator:
    def multiply(self,*args):
        value = 1
        for i in args:
            value *=i
        return value

cal1 = Calculator()
cal1.multiply(3,3,7)
```

```python
# 41. Banking Transaction System
#     - Create a base class Transaction.
#     - Create Deposit, Withdrawal, and Transfer subclasses.
#     - Implement a common execute() method.
#     - Override execute() in each subclass.
#     - Process multiple transactions using a loop.

class Transaction:
    def execute(self):
        print(f"Transactions executed")

class Deposit(Transaction):
    def execute(self):
        print(f"Transactions executed: Deposit")

class Withdrawal(Transaction):
    def execute(self):
        print(f"Transactions executed: Withdrawal")

class Transfer(Transaction):
    def execute(self):
        print(f"Transactions executed: Transfer")

transactions = [Deposit(),Withdrawal(),Transfer()]

for transaction in transactions:
    transaction.execute()
```
