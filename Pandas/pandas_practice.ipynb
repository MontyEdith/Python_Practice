```python
# 62. Create Grades Using apply()
# Create a custom function that assigns:
# - A for marks ≥ 90
# - B for marks ≥ 75
# - C for marks ≥ 60
# - D otherwise
# Apply it to the DataFrame.

def grade_assign(a):
    if a >= 90:
        return "A"
    elif a >= 75:
        return "B"
    elif a >= 60:
        return "C"
    else:
        return "D"


data = {
    "Chemistry" : [12,83,94,76,67],
    "Physics" : [96,77,68,89,30],
    "Maths" : [98,76,54,91,87]
}

student = pd.DataFrame(data,index=["Rishabh","Clark","Frank","Groot","Latoor"])

student["Chemistry_Grade"] = student["Chemistry"].apply(grade_assign)
student["Physics_Grade"] = student["Physics"].apply(grade_assign)
student["Maths_Grade"] = student["Maths"].apply(grade_assign)

print(student)
```

```python
# 94. Encode Categories in Different Ways
# Convert a Gender column into binary values using:
# - .cat.codes
# - .replace()
# - .apply()
# Compare the results.

df = pd.DataFrame({
    "Name": ["Rishabh", "Monty", "EniMeM", "Rahul", "Priya"],
    "Gender": ["Male", "Female", "Male", "Male", "Female"]
})

# Method 1: Using .cat.codes
df["Code_cat"] = df["Gender"].astype("category").cat.codes

# Method 2: Using .replace()
df["Code_replace"] = df["Gender"].replace({"Male": 1,"Female": 0})

# Method 3: Using .apply()
df["Code_apply"] = df["Gender"].apply(lambda x: 1 if x == "Male" else 0)

print(df)
```
