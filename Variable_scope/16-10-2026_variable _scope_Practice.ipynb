```python
# [ ] 25. Build a simple bank balance

# Create a global balance = 5000. Write deposit(amount) and withdraw(amount) functions that update the balance using global. Prevent withdrawals that exceed the balance.

balance = 5000

def deposit(amount):
    global balance
    balance += amount
    print(f"After depositing {amount}, your current balance is {balance}")

def withdraw(amount):
    global balance
    if balance < amount:
        print("You don't have enough balance")
    else:
        balance -= amount
        print(f"After withdrawing {amount}, your current balance is {balance}")

deposit(1000)

withdraw(4000)


```

```python
# [ ] 26. Track function calls

# Create a global variable calls = 0. Write a function track_calls() that increases it every time the function is called. Display the total after seven calls.

calls = 0

def track_calls():
    global calls
    calls += 1
    return calls

track_calls()
track_calls()
track_calls()
track_calls()
track_calls()
track_calls()

total_calls = track_calls()
print(total_calls)
```

```python
# [ ] 28. Use global and local variables together

# Create global price = 1000 and discount = 10. Inside a function, calculate the discounted price using local variables without changing the globals.

price = 1000
discount = 10

def cal_discount():
    global price, discount
    discounted_price = price - discount
    print(discounted_price)

cal_discount()
```

```python
# [ ] 29. Mini project: Score tracker

# Create a global score = 0. Implement add_score(points), deduct_score(points), and display_score(). Ensure the score never becomes negative.


score = 0

def add_score(points):
    global score
    score += points

def deduct_score(points):
    global score
    if points > score:
        score = 0
    else:
        score -= points

def display_score():
    global score
    print(f"Final score is {score}")


add_score(100)

deduct_score(20)

display_score()
```
